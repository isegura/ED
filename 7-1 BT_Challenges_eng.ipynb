## 📚 Implementation Summary

The following cell gathers in one place the core implementation covered during the lab: `BinaryNode`, `BinaryTree`, the four traversals (`preorder`, `inorder`, `postorder`, and `levelorder`), and the recursive functions `size` and `height`.

Remember that you can call these functions passing `tree.root` to process the complete tree or with any other node to process only the subtree rooted at that node.


````python
# ============================================================
# SUMMARY OF BINARY TREE ADT
# BinaryNode, BinaryTree, traversals, size and height
# ============================================================

class BinaryNode:
    def __init__(self, e):
        self.element = e
        self.left = None
        self.right = None


class BinaryTree:
    def __init__(self):
        self.root = None


# ------------------------------------------------------------
# DEPTH-FIRST TRAVERSALS
# ------------------------------------------------------------

def preorder(node):
    """root - left - right"""
    if node is None:
        return []

    return (
        [node.element]
        + preorder(node.left)
        + preorder(node.right)
    )


def inorder(node):
    """left - root - right"""
    if node is None:
        return []

    return (
        inorder(node.left)
        + [node.element]
        + inorder(node.right)
    )


def postorder(node):
    """left - right - root"""
    if node is None:
        return []

    return (
        postorder(node.left)
        + postorder(node.right)
        + [node.element]
    )


# ------------------------------------------------------------
# LEVEL-ORDER TRAVERSAL
# ------------------------------------------------------------

def levelorder(node):
    """Iterative level-order traversal using a FIFO queue."""
    if node is None:
        return []

    queue = [node]
    result = []

    while queue:
        current = queue.pop(0)
        result.append(current.element)

        if current.left is not None:
            queue.append(current.left)

        if current.right is not None:
            queue.append(current.right)

    return result


# ------------------------------------------------------------
# SIZE
# ------------------------------------------------------------

def size(node):
    """Returns the number of nodes in the tree or subtree."""
    if node is None:
        return 0

    return 1 + size(node.left) + size(node.right)


# ------------------------------------------------------------
# HEIGHT
# ------------------------------------------------------------

def height(node):
    """Returns the height of the tree or subtree.

    Convention:
        empty tree -> -1
        leaf       -> 0
    """
    if node is None:
        return -1

    return 1 + max(
        height(node.left),
        height(node.right)
    )
````
