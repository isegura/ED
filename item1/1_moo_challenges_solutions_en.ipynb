# 🏆 Challenge 1. Recognizing Python Types

🎯 **Objective:** Implement `python_type(value)` to return the name of the type of the received value.

🕒 **Estimated time:** 5 minutes

🟢 **Difficulty:** Easy

The function must recognize these types:

`int`, `float`, `bool`, `str`, `list`, `tuple`, `dict`, and `set`.

The result must be a string with the type name.

### 🧠 Before Programming
- What does `type(value)` return?
- How can you obtain only the name of the type?
- Remember that `bool` is also a Python type.


````python
def python_type(value):
    """Return the name of the type of the received value."""
    return type(value).__name__
````

Run the following cell to check your solution.


````python
assert python_type(10) == "int"
assert python_type(3.5) == "float"
assert python_type(True) == "bool"
assert python_type("Ana") == "str"
assert python_type([1, 2]) == "list"
assert python_type((1, 2)) == "tuple"
assert python_type({"a": 1}) == "dict"
assert python_type({1, 2}) == "set"
print("Tests passed")
````

# 🏆 Challenge 6. Implementing the ADT Scoreboard

🎯 **Objective:** Implement an Abstract Data Type using a Python class.

🕒 **Estimated time:** 20 minutes

🟡 **Difficulty:** Medium

### 📋 Statement

In the previous challenge, you identified the attributes and methods of the **ADT Scoreboard**, which represents the score of a football match.

Now implement the `Scoreboard` class in Python.

The class must include:

- A constructor that receives the number of goals scored by the home team and the away team.
- The attributes:
  - `home_goals`
  - `away_goals`
- A method `home_goal()` that increases the home team's goals by one.
- A method `away_goal()` that increases the away team's goals by one.
- A method `score()` that returns a string with the score in the format:

```text
home_goals - away_goals
```

For example:

```python
scoreboard = Scoreboard(1, 0)

scoreboard.home_goal()
scoreboard.away_goal()
scoreboard.away_goal()

print(scoreboard.score())
```

Expected output:

```text
2 - 2
```

### 🧠 Before You Start

- Remember that the constructor must initialize the object's state.
- The `home_goal()` and `away_goal()` methods modify that state.
- The `score()` method does not modify the object; it only returns a representation of the scoreboard.


````python
class Scoreboard:
    """Represent the score of a football match."""

    def __init__(self, home_goals, away_goals):
        self.home_goals = home_goals
        self.away_goals = away_goals

    def home_goal(self):
        self.home_goals += 1

    def away_goal(self):
        self.away_goals += 1

    def score(self):
        return f"{self.home_goals} - {self.away_goals}"
````

Run the following cell to check your solution.


````python
scoreboard = Scoreboard(1, 0)

assert scoreboard.home_goals == 1
assert scoreboard.away_goals == 0

scoreboard.home_goal()
assert scoreboard.home_goals == 2
assert scoreboard.away_goals == 0

scoreboard.away_goal()
scoreboard.away_goal()

assert scoreboard.home_goals == 2
assert scoreboard.away_goals == 2

assert scoreboard.score() == "2 - 2"

print("Tests passed")
````

# 🏆 Challenge 8. Encapsulation

🎯 **Objective:** Implement a `Thermostat` class that ensures the temperature stays within a valid range.

🕒 **Estimated time:** 15 minutes

🟡 **Difficulty:** Medium

The class must include:

- attribute `temperature`
- method `set_temperature(new_temperature)`
- method `get_temperature()`

Only temperatures between `5` and `35` degrees, both included, will be accepted.
`set_temperature` must return `True` if it changes the value and `False` if it rejects it.

### 🧠 Before Programming
- The validation must be inside the method.
- If the value is invalid, the object's state must be preserved.


````python
class Thermostat:
    """Keep a temperature within the allowed range."""

    def __init__(self, temperature):
        self.temperature = temperature

    def set_temperature(self, new_temperature):
        if 5 <= new_temperature <= 35:
            self.temperature = new_temperature
            return True
        return False

    def get_temperature(self):
        return self.temperature
````

Run the following cell to check your solution.


````python
thermostat = Thermostat(20)

assert thermostat.get_temperature() == 20
assert thermostat.set_temperature(25) is True
assert thermostat.get_temperature() == 25

assert thermostat.set_temperature(40) is False
assert thermostat.get_temperature() == 25

assert thermostat.set_temperature(5) is True
assert thermostat.set_temperature(35) is True
print("Tests passed")
````

# 🏆 Challenge 9. Inheritance

🎯 **Objective:** Complete a simple hierarchy formed by `Animal`, `Dog`, and `Cat`.

🕒 **Estimated time:** 12 minutes

🟡 **Difficulty:** Medium

The base class `Animal` will store `name`.

Each subclass must implement `make_sound()`:

- `Dog` returns `"woof"`
- `Cat` returns `"meow"`

### 🧠 Before Programming
- What attribute do the subclasses share?
- Use `super()` to reuse the base class constructor.


````python
class Animal:

    def __init__(self, name):
        self.name = name


class Dog(Animal):

    def __init__(self, name):
        super().__init__(name)

    def make_sound(self):
        return "woof"


class Cat(Animal):

    def __init__(self, name):
        super().__init__(name)

    def make_sound(self):
        return "meow"
````

Run the following cell to check your solution.


````python
dog = Dog("Luna")
cat = Cat("Milo")

assert dog.name == "Luna"
assert cat.name == "Milo"
assert dog.make_sound() == "woof"
assert cat.make_sound() == "meow"
assert isinstance(dog, Animal)
assert isinstance(cat, Animal)
print("Tests passed")
````

# 🏆 Challenge 10. Polymorphism

🎯 **Objective:** Implement a function that works with different objects that share the same operation.

🕒 **Estimated time:** 10 minutes

🟡 **Difficulty:** Medium

Implement `sounds(animals)`.

The function receives a list of objects and returns a list with the result of calling `make_sound()` on each one.

### 🧠 Before Programming
- The function does not need to check whether each object is a dog or a cat.
- It is enough that all of them provide the same method.


````python
def sounds(animals):
    """Return the sound produced by each object in the collection."""
    return [animal.make_sound() for animal in animals]
````

Run the following cell to check your solution.


````python
class Dog:
    def make_sound(self):
        return "woof"

class Cat:
    def make_sound(self):
        return "meow"

class Bird:
    def make_sound(self):
        return "tweet"

animals = [Dog(), Cat(), Bird()]
assert sounds(animals) == ["woof", "meow", "tweet"]
assert sounds([]) == []
print("Tests passed")
````

# 🏆 Challenge 11. Debugging a Class

🎯 **Objective:** Fix the errors in the `Product` class.

🕒 **Estimated time:** 12 minutes

🟡 **Difficulty:** Medium

The class must:

- store `name` and `price`
- allow a percentage discount to be applied
- reject discounts lower than `0` or greater than `100`
- return `True` when the discount is applied and `False` when it is rejected

### 🧠 Before Programming
Look for errors related to:

- use of `self`
- modification of attributes
- data validation


````python
class Product:

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def apply_discount(self, percentage):
        if 0 <= percentage <= 100:
            self.price -= self.price * percentage / 100
            return True
        return False
````

Run the following cell to check your solution.


````python
product = Product("Keyboard", 80)

assert product.name == "Keyboard"
assert product.price == 80

assert product.apply_discount(25) is True
assert product.price == 60

assert product.apply_discount(-10) is False
assert product.price == 60

assert product.apply_discount(120) is False
assert product.price == 60
print("Tests passed")
````

# 🏆 Challenge 12. Implementing the ADT Complex Number

🎯 **Objective:** Implement a simple class to represent complex numbers.

🕒 **Estimated time:** 20 minutes

🔴 **Difficulty:** Difficult

The `ComplexNumber` class must store:

- `real`
- `imaginary`

It must implement:

- `add(other)`, which returns a new `ComplexNumber`
- `subtract(other)`, which returns a new `ComplexNumber`
- `modulus()`, which returns its modulus

### 🧠 Before Programming
For a complex number `a + bi`:

- addition and subtraction are performed component by component;
- the modulus is `(a? + b?) ** 0.5`.


````python
class ComplexNumber:

    def __init__(self, real, imaginary):
        self.real = real
        self.imaginary = imaginary

    def add(self, other):
        return ComplexNumber(
            self.real + other.real,
            self.imaginary + other.imaginary
        )

    def subtract(self, other):
        return ComplexNumber(
            self.real - other.real,
            self.imaginary - other.imaginary
        )

    def modulus(self):
        return (self.real ** 2 + self.imaginary ** 2) ** 0.5
````

Run the following cell to check your solution.


````python
z1 = ComplexNumber(3, 4)
z2 = ComplexNumber(1, -2)

sum_result = z1.add(z2)
difference = z1.subtract(z2)

assert isinstance(sum_result, ComplexNumber)
assert sum_result.real == 4
assert sum_result.imaginary == 2

assert difference.real == 2
assert difference.imaginary == 6

assert z1.modulus() == 5
print("Tests passed")
````

# 🏆 Challenge 13. Modeling a Library

🎯 **Objective:** Represent a simplified version of a loan system using classes.

🕒 **Estimated time:** 25 minutes

🔴 **Difficulty:** Difficult

Implement:

### `Book`

Attributes:

- `isbn`
- `title`
- `available`, initially `True`

### `User`

Attributes:

- `identifier`
- `name`
- `loans`, initially an empty list

Methods:

- `borrow(book)`: borrows the book if it is available
- `return_book(book)`: returns a borrowed book

The methods must return `True` when the operation is performed and `False` otherwise.

### 🧠 Before Programming
- When a book is borrowed, it must no longer be available.
- The same book cannot be borrowed twice at the same time.
- When it is returned, it must be removed from the loans list and become available again.


````python
class Book:

    def __init__(self, isbn, title):
        self.isbn = isbn
        self.title = title
        self.available = True


class User:

    def __init__(self, identifier, name):
        self.identifier = identifier
        self.name = name
        self.loans = []

    def borrow(self, book):
        if book.available:
            book.available = False
            self.loans.append(book)
            return True
        return False

    def return_book(self, book):
        if book in self.loans:
            self.loans.remove(book)
            book.available = True
            return True
        return False
````

Run the following cell to check your solution.


````python
book = Book("978-1", "Data Structures")
ana = User("U01", "Ana")
luis = User("U02", "Luis")

assert book.available is True
assert ana.loans == []

assert ana.borrow(book) is True
assert book.available is False
assert book in ana.loans

assert luis.borrow(book) is False
assert book not in luis.loans

assert ana.return_book(book) is True
assert book.available is True
assert book not in ana.loans

assert ana.return_book(book) is False
print("Tests passed")
````
