# 🏆 Challenge 1. Counting Signs

🎯 **Objective:** Analyze the time and space complexity of the following function.

🕒 **Estimated time:** 8 minutes  
🟢 **Difficulty:** Easy


```python
def sign_summary(values):
    summary = {
        "positives": 0,
        "negatives": 0,
        "zeros": 0,
    }

    for value in values:
        if value > 0:
            summary["positives"] += 1
        elif value < 0:
            summary["negatives"] += 1
        else:
            summary["zeros"] += 1

    return summary
```

### 🔍 Questions

1. What does \(n\) represent in this algorithm?
2. How many times is the loop executed?
3. What is its time complexity?
4. What is its auxiliary space complexity?
5. Do the best and worst cases change?

### ✍️ Answer

- **\(n\):** represents the number of elements in the list `values`.
- **Number of iterations:** the loop is executed \(n\) times.
- **Time complexity:** \(O(n)\).
- **Auxiliary space complexity:** \(O(1)\).
- **Best and worst cases:** they do not change; in both cases all elements are traversed.

**Justification:** in each iteration, a constant number of operations is performed. The dictionary `summary` always contains three entries, regardless of the size of the list.


# 🏆 Challenge 2. Cumulative Totals

🎯 **Objective:** Analyze the time and space complexity of the following function.

🕒 **Estimated time:** 8 minutes  
🟢 **Difficulty:** Easy


```python
def cumulative_totals(values):
    result = []
    total = 0

    for value in values:
        total += value
        result.append(total)

    return result
```

### 🔍 Questions

1. How many elements does `result` contain at the end?
2. What is the time complexity?
3. What is the space complexity?
4. Could the space complexity be reduced without changing what the function returns?

### ✍️ Answer

- **Elements in `result`:** \(n\), one for each element in `values`.
- **Time complexity:** \(O(n)\).
- **Space complexity:** \(O(n)\).
- **Can it be reduced?:** no, if the function must still return a list with all cumulative totals.

**Justification:** the algorithm traverses the input once. The returned list grows until it contains \(n\) elements, so it requires linear space.


# 🏆 Challenge 3. Removing Duplicates

🎯 **Objective:** Analyze the time and space complexity of the following function:

🕒 **Estimated time:** 12 minutes  
🟡 **Difficulty:** Medium


```python
def remove_duplicates(values):
    result = []

    for value in values:
        if value not in result:
            result.append(value)

    return result
```

### 🔍 Questions

1. What is the cost of `value not in result` when `result` is a list?
2. What does the number of comparisons performed by that operation depend on?
3. What is the worst case for this algorithm? Justify your answer.
4. What is the time complexity of the algorithm?
5. What is the space complexity of the algorithm?
6. Which Python data structure would you use to reduce the cost of membership checks? How would the time complexity of the algorithm change?

### ✍️ Answer

- **Cost of `value not in result`:** \(O(k)\), where \(k\) is the number of elements currently stored in `result`.
- **Number of comparisons:** it depends on the size of `result` and on the position where the element is found, if it was already included.
- **Worst case:** all values are different. In that case, `result` grows in every iteration and approximately \(0 + 1 + 2 + \cdots + (n-1)\) comparisons are performed.
- **Time complexity:** \(O(n^2)\) in the worst case.
- **Space complexity:** \(O(n)\).
- **Possible improvement:** use a `set` to check whether an element has already appeared. Membership in a set has average cost \(O(1)\), so the complete algorithm would have average time cost \(O(n)\), while keeping \(O(n)\) space.


# 🏆 Challenge 4. Row Sums

🎯 **Objective:** Analyze an algorithm that traverses a rectangular matrix (with dimensions r x c) using nested loops to calculate the sum of the elements in each row.

🕒 **Estimated time:** 12 minutes  
🟡 **Difficulty:** Medium

The following function receives a **matrix** represented as a list of lists of integers, where all rows have the same number of elements. The function returns a new list with the sum of each row.


```python
def row_sums(matrix):
    result = []

    for row in matrix:
        row_total = 0

        for value in row:
            row_total += value

        result.append(row_total)

    return result
```

### 🔍 Questions

Let \(r\) be the number of rows and \(c\) the number of columns in the matrix.

1. How many times is the instruction `row_total += value` executed?
2. Express the time complexity of the algorithm as a function of \(r\) and \(c\).
3. If the matrix is square with size \(n \times n\), what is the time complexity?
4. What is the space complexity of the list `result`?


### ✍️ Answer

- **Number of executions of `row_total += value`:**  
  It is executed once for each element in the matrix, that is, \(r \cdot c\) times.

- **Time complexity for an \(r \times c\) matrix:**  
  \[
  O(r \cdot c)
  \]

- **Time complexity for an \(n \times n\) matrix:**  
  \[
  O(n^2)
  \]

- **Space complexity:**  
  The list `result` stores one sum for each row, so it takes \(O(r)\) space.


# 🏆 Challenge 5. Validating a Rectangular Matrix

🎯 **Objective**: Analyze how the execution time of an algorithm changes depending on the input case.
🕒 **Estimated time:** 10 minutes  
🟡 **Difficulty:** Medium


```python
def is_rectangular(matrix):
    if len(matrix) == 0:
        return True

    length = len(matrix[0])

    for row in matrix:
        if len(row) != length:
            return False

    return True
```

### 🔍 Questions

Let \(n\) be the number of rows.

1. What input produces the best case?
2. What input produces the worst case?
3. What is the time complexity of the best case?
4. What is the time complexity of the worst case?
5. What is the auxiliary space complexity?

### ✍️ Answer

- **Best case:** the matrix is empty, or the first checked row has a different length from the first row. In both cases, the function finishes immediately or after a constant number of checks.
- **Worst case:** all rows have the same length, or the first different row appears at the end. The algorithm must examine the \(n\) rows.
- **Best-case time complexity:** \(O(1)\).
- **Worst-case time complexity:** \(O(n)\).
- **Auxiliary space complexity:** \(O(1)\).

**Justification:** `len(row)` has constant cost for a Python list, and the algorithm does not create any structure whose size depends on \(n\).


# 🏆 Challenge 6. Constant Cost vs. Linear Cost

🎯 **Objective:** Compare two algorithms that solve the same problem.

🕒 **Estimated time:** 10 minutes  
🟢 **Difficulty:** Easy


```python
def iterative_sum(n):
    total = 0

    for i in range(1, n + 1):
        total += i

    return total


def gauss_sum(n):
    return n * (n + 1) // 2
```

### 🔍 Questions

1. What is the time complexity of each function?
2. What is the space complexity of each function?
3. Which one scales better when \(n\) is very large?
4. Which function would you use for \(n = 10^9\)? Justify your answer.

### ✍️ Answer

| Function | Time complexity | Space complexity |
|---|---|---|
| `iterative_sum` | \(O(n)\) | \(O(1)\) |
| `gauss_sum` | \(O(1)\) | \(O(1)\) |

**Conclusion:** `gauss_sum` scales better because it performs a constant number of operations, regardless of the value of \(n\). For \(n = 10^9\), I would use `gauss_sum`, since it avoids executing one billion iterations.


# 🧪 Challenge 7. Empirical Analysis

🎯 **Objective:** Experimentally measure the execution time of an algorithm for different input sizes and plot the results.

🕒 **Estimated time:** 20 minutes  
🟡 **Difficulty:** Medium


```python
def count_even(values):
    count = 0

    for value in values:
        if value % 2 == 0:
            count += 1

    return count
```

### Tasks

1. Generate lists with sizes \(10^1, 10^2, \ldots, 10^7\).
2. Measure the execution time with `time.time()`.
3. Store each pair `(size, time)` in a list called `measurements`.
4. Create a `DataFrame` with the results.
5. Plot **Execution time vs. Input size**.
6. Indicate the trend you observe and compare it with the asymptotic analysis.


```python
import time
import pandas as pd
import matplotlib.pyplot as plt

measurements = []

for exponent in range(1, 8):
    n = 10 ** exponent
    values = list(range(n))

    start = time.time()
    count_even(values)
    end = time.time()

    measurements.append((n, end - start))

df = pd.DataFrame(measurements, columns=["N", "Time (s)"])
df
```

```python
plt.figure(figsize=(8, 5))
plt.plot(df["N"], df["Time (s)"], marker="o")

plt.title("Empirical analysis of count_even")
plt.xlabel("Input size (N)")
plt.ylabel("Execution time (s)")
plt.grid(True)
plt.show()
```

### ✍️ Conclusion

- **Observed trend:** the execution time increases approximately linearly as the list size grows.
- **Expected time complexity:** \(O(n)\).
- **Do the empirical and asymptotic analyses match?:** yes. The function traverses all elements once and performs a constant amount of work per element. Measurements may show small variations due to the system and the measurement mechanism itself.


# 🏆 Challenge 8. True or False

🎯 **Objective:** Detect common incorrect statements about algorithm analysis.

🕒 **Estimated time:** 10 minutes  
🟡 **Difficulty:** Medium

Indicate whether each statement is true or false and briefly justify your answer.

1. Big-O notation tells us exactly how long an algorithm will take.
2. Two \(O(n)\) algorithms can have different execution times.
3. An \(O(1)\) algorithm takes exactly the same number of seconds for every input.
4. Big-O notation depends on the programming language used.
5. For large values of \(n\), an \(O(n)\) algorithm will be more efficient than an \(O(n^2)\) algorithm.
6. Two nested loops necessarily imply \(O(n^2)\) complexity.
7. Space complexity always includes the memory occupied by input parameters.
8. An algorithm with \(O(1)\) time complexity can use more memory than another algorithm with \(O(n)\) complexity.

### ✍️ Answers

| No. | T/F | Justification |
|---:|:---:|---|
| 1 | F | Big-O describes how the cost of an algorithm grows, but it does not provide an exact time in seconds. |
| 2 | T | They may perform different numbers of operations per element and have different constants. |
| 3 | F | \(O(1)\) means that the number of operations does not grow with the input size; the real time may vary. |
| 4 | F | Asymptotic complexity describes the growth of the algorithm and does not depend on the specific language. |
| 5 | T | For sufficiently large inputs, linear growth is smaller than quadratic growth. |
| 6 | F | It depends on how many times each loop is executed. If the inner loop always performs 10 iterations, the total cost is \(O(n)\). |
| 7 | F | In this practice, we consider auxiliary space, which does not include the memory occupied by input parameters. |
| 8 | T | Time complexity and space complexity measure different resources; a constant-time algorithm can allocate a large structure. |
