🕒 **Estimated time:** 8 minutes  
🟢 **Difficulty:** Easy


```python
def sign_summary(values):
    summary = {
        "positives": 0,
        "negatives": 0,
        "zeros": 0,
    }

    for value in values:
        if value > 0:
            summary["positives"] += 1
        elif value < 0:
            summary["negatives"] += 1
        else:
            summary["zeros"] += 1

    return summary
```

# 🏆 Challenge 5. Validating a Rectangular Matrix

🎯 **Objective**: Analyze how the execution time of an algorithm changes depending on the input case.
🕒 **Estimated time:** 10 minutes  
🟡 **Difficulty:** Medium


```python
def is_rectangular(matrix):
    if len(matrix) == 0:
        return True

    length = len(matrix[0])

    for row in matrix:
        if len(row) != length:
            return False

    return True
```

# 🏆 Challenge 6. Constant Cost vs. Linear Cost

🎯 **Objective:** Compare two algorithms that solve the same problem.

🕒 **Estimated time:** 10 minutes  
🟢 **Difficulty:** Easy


```python
def iterative_sum(n):
    total = 0

    for i in range(1, n + 1):
        total += i

    return total


def gauss_sum(n):
    return n * (n + 1) // 2
```
