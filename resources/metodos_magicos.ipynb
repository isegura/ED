# Métodos mágicos en Python

En Python, muchos comportamientos de los objetos están controlados por métodos especiales cuyos nombres empiezan y terminan con dos guiones bajos:

```python
__metodo__
```

A estos métodos se les suele llamar **métodos mágicos** o **dunder methods** (*double underscore methods*).

Normalmente no los invocamos directamente: Python los llama automáticamente cuando utilizamos determinadas operaciones sobre los objetos.

## 1. ¿Por qué existen?

Los métodos mágicos permiten que los objetos definidos por nosotros se comporten de forma natural con la sintaxis de Python.

Por ejemplo, podemos decidir:

- qué muestra `print(objeto)`;
- cómo se calcula `len(objeto)`;
- qué significa `objeto1 == objeto2`;
- qué ocurre con `objeto1 + objeto2`;
- cómo funciona `objeto[indice]`;
- qué significa `elemento in objeto`.

Estos métodos forman parte del **modelo de datos de Python**.

## 2. `__init__`: inicialización de objetos

`__init__` es uno de los métodos mágicos más conocidos.

Python lo ejecuta automáticamente después de crear un objeto para inicializar sus atributos.

````python
class Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

p = Persona("Ana", 20)
print(p.nombre)
print(p.edad)
````

Aunque suele llamarse “constructor” de forma informal, técnicamente `__init__` **inicializa** un objeto que ya ha sido creado.

## 3. `__str__`: representación legible

`__str__` define qué texto se obtiene al utilizar `str(objeto)` o `print(objeto)`.

Debe devolver una cadena.

````python
class Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def __str__(self):
        return f"{self.nombre} ({self.edad} años)"

p = Persona("Ana", 20)
print(p)
print(str(p))
````

## 4. `__repr__`: representación para desarrollo y depuración

`__repr__` pretende proporcionar una representación más técnica del objeto.

Es habitual intentar que sea clara e inequívoca.

````python
class Punto:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self):
        return f"Punto(x={self.x}, y={self.y})"

p = Punto(3, 5)
print(repr(p))
print([p])
````

## 5. `__len__`: definir `len()`

Si una clase implementa `__len__`, podemos utilizar `len(objeto)`.

````python
class Grupo:
    def __init__(self, estudiantes):
        self.estudiantes = estudiantes

    def __len__(self):
        return len(self.estudiantes)

g = Grupo(["Ana", "Luis", "Marta"])
print(len(g))
````

`__len__` debe devolver un entero no negativo.

## 6. Comparaciones: `__eq__`, `__lt__`, ...

Podemos definir cómo se comparan nuestros objetos.

Algunos métodos frecuentes son:

- `__eq__` → `==`
- `__ne__` → `!=`
- `__lt__` → `<`
- `__le__` → `<=`
- `__gt__` → `>`
- `__ge__` → `>=`

````python
class Punto:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, otro):
        return self.x == otro.x and self.y == otro.y

p1 = Punto(2, 3)
p2 = Punto(2, 3)
p3 = Punto(4, 5)
print(p1 == p2)
print(p1 == p3)
````

Sin `__eq__`, dos objetos distintos no se consideran iguales únicamente porque sus atributos tengan los mismos valores.

## 7. Operadores aritméticos: `__add__`, `__sub__`, ...

Los operadores también pueden personalizarse.

Por ejemplo:

- `__add__` → `+`
- `__sub__` → `-`
- `__mul__` → `*`
- `__truediv__` → `/`

````python
class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, otro):
        return Vector2D(self.x + otro.x, self.y + otro.y)

    def __repr__(self):
        return f"Vector2D({self.x}, {self.y})"

v1 = Vector2D(1, 2)
v2 = Vector2D(3, 4)
print(v1 + v2)
````

El operador `+` sigue escribiéndose de forma normal, pero Python llama internamente a `__add__`.

## 8. Acceso mediante corchetes: `__getitem__`

`__getitem__` permite definir qué ocurre al escribir:

```python
objeto[indice]
```

````python
class Coleccion:
    def __init__(self, datos):
        self.datos = datos

    def __getitem__(self, indice):
        return self.datos[indice]

c = Coleccion([10, 20, 30])
print(c[0])
print(c[1])
````

## 9. `__contains__`: operador `in`

Podemos definir el comportamiento del operador `in`.

````python
class Grupo:
    def __init__(self, estudiantes):
        self.estudiantes = estudiantes

    def __contains__(self, nombre):
        return nombre in self.estudiantes

g = Grupo(["Ana", "Luis", "Marta"])
print("Ana" in g)
print("Pedro" in g)
````

## 10. No solemos invocarlos directamente

Una idea importante es que normalmente **no llamamos directamente** a estos métodos.

En lugar de:

```python
objeto.__len__()
```

preferimos:

```python
len(objeto)
```

Y en lugar de:

```python
objeto.__add__(otro)
```

preferimos:

```python
objeto + otro
```

La sintaxis habitual es más clara y expresa mejor la intención del programa.

## 11. Resumen

| Operación | Método mágico |
|---|---|
| Inicializar objeto | `__init__` |
| `str(objeto)` / `print(objeto)` | `__str__` |
| `repr(objeto)` | `__repr__` |
| `len(objeto)` | `__len__` |
| `objeto1 == objeto2` | `__eq__` |
| `objeto1 < objeto2` | `__lt__` |
| `objeto1 + objeto2` | `__add__` |
| `objeto[indice]` | `__getitem__` |
| `elemento in objeto` | `__contains__` |

No es necesario implementar todos estos métodos. Solo aquellos que tengan sentido para el comportamiento de nuestra clase.
