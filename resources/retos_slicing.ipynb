## 1. ¿Qué es el slicing?

El *slicing* permite obtener una parte de una secuencia utilizando la sintaxis:

```python
secuencia[inicio:fin:paso]
```

- `inicio`: posición desde la que empezamos.
- `fin`: posición en la que nos detenemos **sin incluirla**.
- `paso`: incremento entre posiciones.

Cualquiera de los tres valores puede omitirse.

````python
texto = "Estructuras"

print(texto[0:4])   # posiciones 0, 1, 2 y 3
print(texto[4:])    # desde la posición 4 hasta el final
print(texto[:5])    # desde el principio hasta la posición 4
print(texto[:])     # copia completa
````

## 2. Índices negativos

Python permite contar posiciones desde el final utilizando índices negativos.

Por ejemplo, en `"Python"`:

```text
 P  y  t  h  o  n
 0  1  2  3  4  5
-6 -5 -4 -3 -2 -1
```

Esto resulta especialmente útil en combinación con slicing.

````python
texto = "Python"

print(texto[-1])      # último carácter
print(texto[-3:])     # últimos tres caracteres
print(texto[:-2])     # todo excepto los dos últimos
````

## 3. El paso

El tercer valor del slicing indica cada cuántos elementos avanzamos.

```python
secuencia[inicio:fin:paso]
```

Si el paso es negativo, recorremos la secuencia de derecha a izquierda.

````python
texto = "abcdefghij"

print(texto[::2])     # posiciones pares
print(texto[1::2])    # posiciones impares
print(texto[::-1])    # secuencia invertida
print(texto[::-2])    # de derecha a izquierda, saltando de dos en dos
````

## 4. Slicing con listas

El slicing funciona de la misma forma con listas.

````python
numeros = [10, 20, 30, 40, 50, 60, 70]

print(numeros[1:4])
print(numeros[-3:])
print(numeros[::2])
print(numeros[::-1])
````

## Algunas ideas importantes

- El índice `fin` **no se incluye**.
- Si omitimos `inicio`, Python empieza desde el principio.
- Si omitimos `fin`, Python continúa hasta el final.
- Un `paso` de `2` selecciona un elemento de cada dos.
- Un `paso` de `-1` permite invertir una secuencia.
- El slicing no modifica la secuencia original: devuelve una nueva secuencia.
