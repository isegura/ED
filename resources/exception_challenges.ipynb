# Exceptions in Python

When we run a program, situations may occur that prevent normal execution from continuing: trying to divide by zero, converting text that does not represent a number to an integer, accessing a non-existent position in a list, etc.

Python represents these situations using **exceptions**.

In this notebook we will learn:

- what an exception is;
- why handling exceptions is useful;
- how to use `try` and `except`;
- how to handle different types of exceptions;
- what `else` and `finally` are used for;
- how to raise an exception with `raise`.

At the end, you will find several challenges to practise.

## 1. What is an exception?

An **exception** is an error that occurs **during the execution** of a program.

For example:

````python
a = 10
b = 0

# This operation would raise an exception:
# resultado = a / b
````

If we executed the last line, we would get an error similar to:

```text
ZeroDivisionError: division by zero
```

`ZeroDivisionError` is the exception type Python uses when we try to divide by zero.

When an exception occurs and is not handled, normal program execution is interrupted.

## 2. Why do we need to handle exceptions?

In many programs, certain errors may occur for reasons that we cannot completely avoid.

For example, a user may enter text when we expect a number:

````python
texto = "hola"

# This conversion would raise an exception:
# numero = int(texto)
````

In this case, Python would raise:

```text
ValueError: invalid literal for int()
```

It is often simpler and clearer to **try to perform the operation and handle the error if it occurs**.

For this, we use `try` and `except`.

## 3. `try` and `except`

The basic structure is:

```python
try:
    # código que puede producir una excepción
except TipoDeExcepcion:
    # código que se ejecuta si se produce esa excepción
```

The `try` block contains the code we want to execute.

If the exception indicated in `except` occurs, Python stops the `try` block and executes the `except` block.

````python
try:
    numero = int("hola")
    print(numero)
except ValueError:
    print("The text does not represent an integer.")

print("The program continues.")
````

The goal of handling an exception is not to hide errors, but to decide **what the program should do** when an exceptional situation occurs.

## 4. An example with a function

We can use exceptions inside our functions.

For example:

````python
def dividir(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return None

print(dividir(10, 2))
print(dividir(10, 0))
````

In this example, if `b` is `0`, the operation `a / b` raises `ZeroDivisionError`. The function catches the exception and returns `None`.

The decision to return `None`, display a message, or perform another action depends on the problem we are solving.

## 5. Some common exception types

| Exception | Typical situation |
|---|---|
| `ZeroDivisionError` | Division by zero |
| `ValueError` | Invalid value for an operation |
| `TypeError` | Operation applied to an incorrect type |
| `IndexError` | Non-existent sequence position |
| `KeyError` | Non-existent dictionary key |

````python
# ValueError
try:
    int("Python")
except ValueError:
    print("Cannot convert 'Python' to an integer.")

# IndexError
try:
    lista = [10, 20, 30]
    print(lista[10])
except IndexError:
    print("The position does not exist.")

# TypeError
try:
    resultado = "10" + 5
except TypeError:
    print("A string and an integer cannot be added directly.")
````

## 6. Multiple `except` blocks

The same `try` block may produce different types of exceptions.

We can use several `except` blocks to handle them differently:

````python
def dividir_textos(a, b):
    try:
        x = int(a)
        y = int(b)
        return x / y
    except ValueError:
        return "One of the values is not an integer."
    except ZeroDivisionError:
        return "Cannot divide by zero."

print(dividir_textos("10", "2"))
print(dividir_textos("hola", "2"))
print(dividir_textos("10", "0"))
````

It is advisable to catch **specific exceptions** when we know which errors we expect.

For example, it is usually preferable to use:

```python
except ValueError:
```

rather than simply using:

```python
except:
```

because this avoids hiding other errors that could indicate a problem in our code.

## 7. The `else` block

The `else` and `finally` blocks are **optional**. They are not required in order to handle an exception.

The `else` block is executed **only if the `try` block finishes without raising any exception**.

For example:

````python
try:
    numero = int("25")
except ValueError:
    print("It is not an integer.")
else:
    print("The conversion was successful.")
    print("Number:", numero)
````

In this example:

- if `int("25")` succeeds, `else` is executed;
- if `ValueError` occurs, `except` is executed and `else` is **not** executed.

The `else` block is useful for code that should run only when the operation in `try` finishes successfully.

It also helps keep inside `try` only the statements that can actually raise the exception we want to handle.

## 8. The `finally` block

The `finally` block is also **optional**, but it behaves differently from `else`.

`finally` is executed **always**, whether the `try` block succeeds or an exception occurs.

For example:

````python
try:
    numero = int("abc")
except ValueError:
    print("An error occurred.")
finally:
    print("End of the operation.")
````

In this case, `ValueError` occurs, so `except` is executed. After that, `finally` is also executed.

`finally` is commonly used for tasks that must always be performed at the end of an operation, such as closing a file or releasing a resource.

### Difference between `else` and `finally`

| Block | When is it executed? |
|---|---|
| `try` | Python always tries to execute it |
| `except` | Only if the corresponding exception occurs |
| `else` | Only if **no** exception occurs |
| `finally` | **Always** |

We may use only `try` and `except`, add `else`, add `finally`, or use both.

For example:

```python
try:
    numero = int("10")
except ValueError:
    print("Conversion error")
else:
    print("Successful conversion:", numero)
finally:
    print("End of the process")
```

If the conversion succeeds, `try`, `else`, and `finally` are executed.  
If it fails, `try`, `except`, and `finally` are executed.

## 9. Raising exceptions with `raise`

So far, we have seen exceptions raised automatically by Python.

We can also raise an exception ourselves using `raise`.

This is useful when we want to indicate that a value received by a function is invalid.

````python
def calcular_media(total, numero_elementos):
    if numero_elementos <= 0:
        raise ValueError("The number of elements must be greater than zero.")

    return total / numero_elementos

print(calcular_media(20, 4))

# This call would raise an exception:
# calcular_media(20, 0)
````

## 10. When should we use exceptions?

Exceptions are especially useful when:

- an operation may fail even if the code is correct;
- we are processing data that may contain invalid values;
- we want a function to clearly indicate that it cannot perform an operation;
- we want to prevent an expected error from abruptly terminating the program.

We should not use exceptions indiscriminately as a replacement for normal control structures.

For example, if we want to check whether a number is positive, it is usually more natural to use an `if`.

## 11. Summary

The general structure may include:

```python
try:
    ...
except TipoDeExcepcion:
    ...
else:
    ...
finally:
    ...
```

It is not necessary to use all the blocks every time.

Además:

```python
raise TipoDeExcepcion("mensaje")
```

allows you to explicitly raise an exception.

### Important ideas

- An exception represents an abnormal situation during execution.
- If it is not caught, it normally interrupts program execution.
- `try` contains the code that may fail.
- `except` allows an exception to be handled.
- It is preferable to catch specific exception types.
- `else` executes if no exception occurred.
- `finally` always executes.
- `raise` allows you to explicitly raise an exception.
