# Working with strings (`str`) in Python

In this notebook we will review some of the most commonly used operations and methods for working with strings in Python.

After the examples, you will find exercises to practise.

## 1. Strings are sequences

A string (`str`) is a sequence of characters. We can:

- Check its length with `len()`.
- Access characters using indexes.
- Traverse it with a loop.
- Concatenate strings with `+`.
- Check whether one text appears inside another with `in`.

Strings are **immutable**: their characters cannot be modified directly.

````python
texto = "Estructuras de Datos"

print(len(texto))
print(texto[0])
print("Datos" in texto)
print(texto + " - Python")
````

## 2. Uppercase and lowercase

Some common methods are:

- `lower()`: converts to lowercase.
- `upper()`: converts to uppercase.
- `capitalize()`: makes the first letter uppercase.
- `title()`: makes the first letter of each word uppercase.

These methods return a **new string**.

````python
texto = "estructuras de DATOS"

print(texto.lower())
print(texto.upper())
print(texto.capitalize())
print(texto.title())
````

## 3. Removing spaces

- `strip()` removes spaces at the beginning and end.
- `lstrip()` removes spaces at the beginning.
- `rstrip()` removes spaces at the end.

````python
texto = "   Python   "

print(repr(texto.strip()))
print(repr(texto.lstrip()))
print(repr(texto.rstrip()))
````

## 4. Searching and counting

Useful methods:

- `find(substring)`: returns the first position where it appears. If it does not appear, it returns `-1`.
- `count(substring)`: counts how many times it appears.
- `startswith(prefix)`: checks whether the string starts with a given text.
- `endswith(suffix)`: checks whether the string ends with a given text.

````python
texto = "programar en python"

print(texto.find("python"))
print(texto.find("java"))
print(texto.count("r"))
print(texto.startswith("pro"))
print(texto.endswith("python"))
````

## 5. Replacing text

`replace(old, new)` returns a new string where occurrences of one text are replaced by another.

````python
texto = "Me gusta Java"

nuevo = texto.replace("Java", "Python")

print(texto)
print(nuevo)
````

## 6. Splitting and joining

`split()` allows you to split a string and obtain a list.

`join()` performs the opposite operation: it joins several strings using a separator.

````python
frase = "uno dos tres"
palabras = frase.split()

print(palabras)
print("-".join(palabras))

datos = "Ana;20;Madrid"
print(datos.split(";"))
````

## 7. Content checks

Some methods return `True` or `False` depending on the content:

- `isalpha()`: letters only.
- `isdigit()`: digits only.
- `isalnum()`: letters or digits.
- `isspace()`: whitespace only.

````python
print("Python".isalpha())
print("12345".isdigit())
print("Aula123".isalnum())
print("   ".isspace())
````

## Some important ideas

- `str` methods normally **do not modify** the original string.
- Python is case-sensitive.
- `find()` returns `-1` if the text is not found.
- `split()` returns a list.
- `join()` is called on the string that will act as the separator.
- It is common to combine several methods to solve a problem.
