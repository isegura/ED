# Excepciones en Python

Cuando ejecutamos un programa pueden producirse situaciones que impiden continuar con la ejecución normal: intentar dividir entre cero, convertir a entero un texto que no representa un número, acceder a una posición inexistente de una lista, etc.

Python representa estas situaciones mediante **excepciones**.

En este notebook aprenderemos:

- qué es una excepción;
- por qué es útil gestionarlas;
- cómo utilizar `try` y `except`;
- cómo tratar distintos tipos de excepciones;
- para qué sirven `else` y `finally`;
- cómo generar una excepción con `raise`.

Al final encontrarás varios retos para practicar.

## 1. ¿Qué es una excepción?

Una **excepción** es un error que se produce **durante la ejecución** de un programa.

Por ejemplo:

````python
a = 10
b = 0

# Esta operación produciría una excepción:
# resultado = a / b
````

Si ejecutásemos la última línea obtendríamos un error similar a:

```text
ZeroDivisionError: division by zero
```

`ZeroDivisionError` es el tipo de excepción que Python utiliza cuando intentamos dividir entre cero.

Cuando se produce una excepción y no se gestiona, la ejecución normal del programa se interrumpe.

## 2. ¿Por qué necesitamos gestionar excepciones?

En muchos programas, ciertos errores pueden producirse por causas que no podemos evitar completamente.

Por ejemplo, un usuario puede introducir texto cuando esperamos un número:

````python
texto = "hola"

# Esta conversión produciría una excepción:
# numero = int(texto)
````

En este caso Python generaría:

```text
ValueError: invalid literal for int()
```

A menudo es más sencillo y claro **intentar realizar la operación y gestionar el error si se produce**.

Para ello utilizamos `try` y `except`.

## 3. `try` y `except`

La estructura básica es:

```python
try:
    # código que puede producir una excepción
except TipoDeExcepcion:
    # código que se ejecuta si se produce esa excepción
```

El bloque `try` contiene el código que queremos ejecutar.

Si se produce la excepción indicada en `except`, Python interrumpe el bloque `try` y ejecuta el bloque `except`.

````python
try:
    numero = int("hola")
    print(numero)
except ValueError:
    print("El texto no representa un número entero.")

print("El programa continúa.")
````

El objetivo de gestionar una excepción no es ocultar los errores, sino decidir **qué debe hacer el programa** cuando se produce una situación excepcional.

## 4. Un ejemplo con una función

Podemos utilizar excepciones dentro de nuestras funciones.

Por ejemplo:

````python
def dividir(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return None

print(dividir(10, 2))
print(dividir(10, 0))
````

En este ejemplo, si `b` vale `0`, la operación `a / b` genera `ZeroDivisionError`. La función captura la excepción y devuelve `None`.

La decisión de devolver `None`, mostrar un mensaje o realizar otra acción dependerá del problema que estemos resolviendo.

## 5. Algunos tipos de excepciones frecuentes

| Excepción | Situación típica |
|---|---|
| `ZeroDivisionError` | División entre cero |
| `ValueError` | Valor no válido para una operación |
| `TypeError` | Operación aplicada a un tipo incorrecto |
| `IndexError` | Posición inexistente de una secuencia |
| `KeyError` | Clave inexistente de un diccionario |

````python
# ValueError
try:
    int("Python")
except ValueError:
    print("No se puede convertir 'Python' a entero.")

# IndexError
try:
    lista = [10, 20, 30]
    print(lista[10])
except IndexError:
    print("La posición no existe.")

# TypeError
try:
    resultado = "10" + 5
except TypeError:
    print("No se pueden sumar directamente una cadena y un entero.")
````

## 6. Varios bloques `except`

Un mismo bloque `try` puede producir distintos tipos de excepciones.

Podemos utilizar varios bloques `except` para tratarlas de forma diferente:

````python
def dividir_textos(a, b):
    try:
        x = int(a)
        y = int(b)
        return x / y
    except ValueError:
        return "Alguno de los valores no es un número entero."
    except ZeroDivisionError:
        return "No se puede dividir entre cero."

print(dividir_textos("10", "2"))
print(dividir_textos("hola", "2"))
print(dividir_textos("10", "0"))
````

Es recomendable capturar **excepciones concretas** cuando sabemos qué errores esperamos.

Por ejemplo, suele ser preferible:

```python
except ValueError:
```

que utilizar simplemente:

```python
except:
```

porque así evitamos ocultar otros errores que podrían indicar un problema en nuestro código.

## 7. El bloque `else`

Los bloques `else` y `finally` son **opcionales**. No es necesario utilizarlos para gestionar una excepción.

El bloque `else` se ejecuta **solo si el bloque `try` termina sin producir ninguna excepción**.

Por ejemplo:

````python
try:
    numero = int("25")
except ValueError:
    print("No es un entero.")
else:
    print("La conversión se realizó correctamente.")
    print("Número:", numero)
````

En este ejemplo:

- si `int("25")` funciona correctamente, se ejecuta `else`;
- si se produce `ValueError`, se ejecuta `except` y **no** se ejecuta `else`.

El bloque `else` resulta útil para colocar código que solo debe ejecutarse cuando la operación del `try` ha terminado correctamente.

Además, permite mantener dentro del `try` únicamente las instrucciones que realmente pueden producir la excepción que queremos gestionar.

## 8. El bloque `finally`

El bloque `finally` también es **opcional**, pero su comportamiento es distinto al de `else`.

`finally` se ejecuta **siempre**, tanto si el bloque `try` termina correctamente como si se produce una excepción.

Por ejemplo:

````python
try:
    numero = int("abc")
except ValueError:
    print("Se ha producido un error.")
finally:
    print("Fin de la operación.")
````

En este caso se produce `ValueError`, por lo que se ejecuta `except`. Después también se ejecuta `finally`.

`finally` suele utilizarse para realizar tareas que deben ejecutarse siempre al terminar una operación, por ejemplo cerrar un fichero o liberar un recurso.

### Diferencia entre `else` y `finally`

| Bloque | ¿Cuándo se ejecuta? |
|---|---|
| `try` | Siempre se intenta ejecutar |
| `except` | Solo si se produce la excepción correspondiente |
| `else` | Solo si **no** se produce ninguna excepción |
| `finally` | **Siempre** |

Podemos utilizar solo `try` y `except`, añadir `else`, añadir `finally`, o utilizar ambos.

Por ejemplo:

```python
try:
    numero = int("10")
except ValueError:
    print("Error en la conversión")
else:
    print("Conversión correcta:", numero)
finally:
    print("Fin del proceso")
```

Si la conversión es correcta, se ejecutan `try`, `else` y `finally`.  
Si falla, se ejecutan `try`, `except` y `finally`.

## 9. Generar excepciones con `raise`

Hasta ahora hemos visto excepciones generadas automáticamente por Python.

También podemos generar una excepción nosotros mismos utilizando `raise`.

Esto es útil cuando queremos indicar que un valor recibido por una función no es válido.

````python
def calcular_media(total, numero_elementos):
    if numero_elementos <= 0:
        raise ValueError("El número de elementos debe ser mayor que cero.")

    return total / numero_elementos

print(calcular_media(20, 4))

# Esta llamada generaría una excepción:
# calcular_media(20, 0)
````

## 10. ¿Cuándo debemos utilizar excepciones?

Las excepciones son especialmente útiles cuando:

- una operación puede fallar aunque el código sea correcto;
- estamos procesando datos que pueden contener valores no válidos;
- queremos que una función indique claramente que no puede realizar una operación;
- queremos evitar que un error esperado termine bruscamente el programa.

No debemos utilizar excepciones para sustituir indiscriminadamente a las estructuras de control normales.

Por ejemplo, si queremos comprobar si un número es positivo, normalmente es más natural utilizar un `if`.

## 11. Resumen

La estructura general puede incluir:

```python
try:
    ...
except TipoDeExcepcion:
    ...
else:
    ...
finally:
    ...
```

No es necesario utilizar siempre todos los bloques.

Además:

```python
raise TipoDeExcepcion("mensaje")
```

permite generar explícitamente una excepción.

### Ideas importantes

- Una excepción representa una situación anómala durante la ejecución.
- Si no se captura, normalmente interrumpe la ejecución del programa.
- `try` contiene el código que puede fallar.
- `except` permite tratar una excepción.
- Es preferible capturar tipos de excepción concretos.
- `else` se ejecuta si no hubo ninguna excepción.
- `finally` se ejecuta siempre.
- `raise` permite generar explícitamente una excepción.
