# Retos: cadenas (`str`) en Python

En este notebook repasaremos algunas de las operaciones y métodos más utilizados para trabajar con cadenas de texto en Python.

Después de los ejemplos encontrarás ejercicios para practicar.

## 1. Las cadenas son secuencias

Una cadena (`str`) es una secuencia de caracteres. Podemos:

- Consultar su longitud con `len()`.
- Acceder a caracteres mediante índices.
- Recorrerla con un bucle.
- Concatenar cadenas con `+`.
- Comprobar si un texto aparece dentro de otro con `in`.

Las cadenas son **inmutables**: sus caracteres no pueden modificarse directamente.

````python
texto = "Estructuras de Datos"

print(len(texto))
print(texto[0])
print("Datos" in texto)
print(texto + " - Python")
````

## 2. Mayúsculas y minúsculas

Algunos métodos frecuentes son:

- `lower()`: convierte a minúsculas.
- `upper()`: convierte a mayúsculas.
- `capitalize()`: primera letra en mayúscula.
- `title()`: primera letra de cada palabra en mayúscula.

Estos métodos devuelven una **nueva cadena**.

````python
texto = "estructuras de DATOS"

print(texto.lower())
print(texto.upper())
print(texto.capitalize())
print(texto.title())
````

## 3. Eliminar espacios

- `strip()` elimina espacios al principio y al final.
- `lstrip()` elimina los del principio.
- `rstrip()` elimina los del final.

````python
texto = "   Python   "

print(repr(texto.strip()))
print(repr(texto.lstrip()))
print(repr(texto.rstrip()))
````

## 4. Buscar y contar

Métodos útiles:

- `find(subcadena)`: devuelve la primera posición donde aparece. Si no aparece, devuelve `-1`.
- `count(subcadena)`: cuenta cuántas veces aparece.
- `startswith(prefijo)`: comprueba si comienza por un texto.
- `endswith(sufijo)`: comprueba si termina por un texto.

````python
texto = "programar en python"

print(texto.find("python"))
print(texto.find("java"))
print(texto.count("r"))
print(texto.startswith("pro"))
print(texto.endswith("python"))
````

## 5. Sustituir texto

`replace(antiguo, nuevo)` devuelve una nueva cadena donde se sustituyen las apariciones de un texto por otro.

````python
texto = "Me gusta Java"

nuevo = texto.replace("Java", "Python")

print(texto)
print(nuevo)
````

## 6. Separar y unir

`split()` permite dividir una cadena y obtener una lista.

`join()` realiza la operación contraria: une varias cadenas utilizando un separador.

````python
frase = "uno dos tres"
palabras = frase.split()

print(palabras)
print("-".join(palabras))

datos = "Ana;20;Madrid"
print(datos.split(";"))
````

## 7. Comprobaciones sobre el contenido

Algunos métodos devuelven `True` o `False` dependiendo del contenido:

- `isalpha()`: solo letras.
- `isdigit()`: solo dígitos.
- `isalnum()`: letras o dígitos.
- `isspace()`: solo espacios en blanco.

````python
print("Python".isalpha())
print("12345".isdigit())
print("Aula123".isalnum())
print("   ".isspace())
````

## Algunas ideas importantes

- Los métodos de `str` normalmente **no modifican** la cadena original.
- Python distingue entre mayúsculas y minúsculas.
- `find()` devuelve `-1` si no encuentra el texto.
- `split()` devuelve una lista.
- `join()` se invoca sobre la cadena que actuará como separador.
- Es habitual combinar varios métodos para resolver un problema.
