## 1. What is slicing?

*Slicing* allows you to obtain part of a sequence using the syntax:

```python
sequence[start:stop:step]
```

- `start`: position where we begin.
- `stop`: position where we stop, **not including it**.
- `step`: increment between positions.

Any of the three values can be omitted.

````python
texto = "Estructuras"

print(texto[0:4])   # positions 0, 1, 2 and 3
print(texto[4:])    # from position 4 to the end
print(texto[:5])    # from the beginning to position 4
print(texto[:])     # full copy
````

## 2. Negative indexes

Python allows positions to be counted from the end using negative indexes.

Por ejemplo, en `"Python"`:

```text
 P  y  t  h  o  n
 0  1  2  3  4  5
-6 -5 -4 -3 -2 -1
```

This is especially useful when combined with slicing.

````python
texto = "Python"

print(texto[-1])      # last character
print(texto[-3:])     # last three characters
print(texto[:-2])     # everything except the last two
````

## 3. The step

The third slicing value indicates how many positions we move forward each time.

```python
sequence[start:stop:step]
```

If the step is negative, we traverse the sequence from right to left.

````python
texto = "abcdefghij"

print(texto[::2])     # even positions
print(texto[1::2])    # odd positions
print(texto[::-1])    # reversed sequence
print(texto[::-2])    # from right to left, taking every other element
````

## 4. Slicing with lists

Slicing works in the same way with lists.

````python
numeros = [10, 20, 30, 40, 50, 60, 70]

print(numeros[1:4])
print(numeros[-3:])
print(numeros[::2])
print(numeros[::-1])
````

## Some important ideas

- The `stop` index is **not included**.
- If `start` is omitted, Python starts from the beginning.
- If `stop` is omitted, Python continues to the end.
- A `step` of `2` selects every other element.
- A `step` of `-1` reverses a sequence.
- Slicing does not modify the original sequence: it returns a new sequence.
