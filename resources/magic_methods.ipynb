# Magic methods in Python

In Python, many object behaviors are controlled by special methods whose names begin and end with two underscores:

```python
__method__
```

These are commonly called **magic methods** or **dunder methods** (*double underscore methods*).

We usually do not call them directly. Python invokes them automatically when certain operations are performed on objects.

## 1. Why do they exist?

Magic methods allow our own objects to work naturally with Python syntax.

For example, we can define:

- what `print(object)` displays;
- how `len(object)` is computed;
- what `object1 == object2` means;
- what happens with `object1 + object2`;
- how `object[index]` behaves;
- what `element in object` means.

These methods are part of Python's **data model**.

## 2. `__init__`: object initialization

`__init__` is one of the best-known magic methods.

Python calls it automatically after an object has been created to initialize its attributes.

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p = Person("Ana", 20)
print(p.name)
print(p.age)
````

Although it is often informally called a constructor, technically `__init__` **initializes** an object that has already been created.

## 3. `__str__`: user-friendly representation

`__str__` defines the text returned by `str(object)` or displayed by `print(object)`.

It must return a string.

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name} ({self.age} years old)"

p = Person("Ana", 20)
print(p)
print(str(p))
````

## 4. `__repr__`: representation for development and debugging

`__repr__` is intended to provide a more technical representation of an object.

It is often designed to be clear and unambiguous.

````python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self):
        return f"Point(x={self.x}, y={self.y})"

p = Point(3, 5)
print(repr(p))
print([p])
````

## 5. `__len__`: defining `len()`

If a class implements `__len__`, we can use `len(object)`.

````python
class Group:
    def __init__(self, students):
        self.students = students

    def __len__(self):
        return len(self.students)

g = Group(["Ana", "Luis", "Marta"])
print(len(g))
````

`__len__` must return a non-negative integer.

## 6. Comparisons: `__eq__`, `__lt__`, ...

We can define how our objects are compared.

Some common methods are:

- `__eq__` → `==`
- `__ne__` → `!=`
- `__lt__` → `<`
- `__le__` → `<=`
- `__gt__` → `>`
- `__ge__` → `>=`

````python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

p1 = Point(2, 3)
p2 = Point(2, 3)
p3 = Point(4, 5)
print(p1 == p2)
print(p1 == p3)
````

Without `__eq__`, two different objects are not considered equal simply because their attributes have the same values.

## 7. Arithmetic operators: `__add__`, `__sub__`, ...

Operators can also be customized.

For example:

- `__add__` → `+`
- `__sub__` → `-`
- `__mul__` → `*`
- `__truediv__` → `/`

````python
class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector2D(self.x + other.x, self.y + other.y)

    def __repr__(self):
        return f"Vector2D({self.x}, {self.y})"

v1 = Vector2D(1, 2)
v2 = Vector2D(3, 4)
print(v1 + v2)
````

The `+` operator is written normally, but Python internally calls `__add__`.

## 8. Square bracket access: `__getitem__`

`__getitem__` lets us define what happens when we write:

```python
object[index]
```

````python
class Collection:
    def __init__(self, data):
        self.data = data

    def __getitem__(self, index):
        return self.data[index]

c = Collection([10, 20, 30])
print(c[0])
print(c[1])
````

## 9. `__contains__`: the `in` operator

We can define the behavior of the `in` operator.

````python
class Group:
    def __init__(self, students):
        self.students = students

    def __contains__(self, name):
        return name in self.students

g = Group(["Ana", "Luis", "Marta"])
print("Ana" in g)
print("Pedro" in g)
````

## 10. We normally do not call them directly

An important idea is that we normally **do not call these methods directly**.

Instead of:

```python
object.__len__()
```

we prefer:

```python
len(object)
```

And instead of:

```python
object.__add__(other)
```

we prefer:

```python
object + other
```

Normal syntax is clearer and better expresses the intention of the program.

## 11. Summary

| Operation | Magic method |
|---|---|
| Initialize object | `__init__` |
| `str(object)` / `print(object)` | `__str__` |
| `repr(object)` | `__repr__` |
| `len(object)` | `__len__` |
| `object1 == object2` | `__eq__` |
| `object1 < object2` | `__lt__` |
| `object1 + object2` | `__add__` |
| `object[index]` | `__getitem__` |
| `element in object` | `__contains__` |

You do not need to implement all of these methods. Only implement those that make sense for the behavior of your class.
