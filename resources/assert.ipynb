# Using `assert` in Python

## Objective

In this notebook, you will learn what the `assert` statement is, how it works, and how to use it to automatically check that a function returns the expected result.


## What Is `assert`?

The syntax is:

```python
assert condition
```

- If the condition is **true**, the program continues.
- If the condition is **false**, Python raises an **`AssertionError`**.


````
# Simple examples

assert 2 + 3 == 5
assert 10 > 5
assert "hello".upper() == "HELLO"

print("All checks are correct.")
````

## What Happens When the Condition Is False?

The following cell produces an `AssertionError` because the obtained result does not match the expected one.


````
def square_v1(x):
    return x * x + 1

# This cell will raise an exception
assert square_v1(4) == 16
````

### Why Does the Error Occur?

The function returns `17`, but the `assert` expects `16`.

That is, Python checks:

```python
17 == 16
```

Since this condition is false, an **`AssertionError`** is raised.


## Using `assert` to Test Functions

Once a function has been implemented, we can write several tests to automatically verify that it works correctly.


````
def square(x):
    return x * x

assert square(0) == 0      # Simple case
assert square(4) == 16     # Positive number
assert square(-3) == 9     # Negative number

print("Tests passed.")
````

## Summary

- `assert` allows you to automatically check that a condition is true.
- If the condition holds, the program continues.
- If it does not hold, Python raises an `AssertionError`.
- It is a very useful tool for verifying that our functions produce the expected results.
