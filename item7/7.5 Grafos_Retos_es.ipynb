> ## 📌 Reglas de la práctica
>
> - Trabajaremos con la clase `Graph` y el atributo interno `self._vertices` del pseudocódigo del tema.
> - `self._vertices[v]` almacena los vértices adyacentes **salientes** desde `v`.
> - Cada conexión se representa mediante un objeto `AdjacentVertex(vertex, weight)`; **no es una tupla**.
> - En grafos no ponderados se usa `weight = 1` por defecto.
> - En un grafo no dirigido, una arista `u-v` debe almacenarse en ambos sentidos.
> - Cuando un método de consulta recibe un vértice inexistente, devolverá un resultado neutro (`[]`, `0` o `False`, según el método) en lugar de modificar el grafo.


## Representación utilizada

El tema representa el grafo mediante un diccionario:

```python
{
    'A': [AdjacentVertex('B', 5), AdjacentVertex('C', 1)],
    'B': [AdjacentVertex('A', 5)],
    'C': [AdjacentVertex('A', 1)]
}
```

La clave es el vértice origen. La lista asociada contiene sus destinos y pesos. Esta representación permite trabajar de forma natural con grafos dirigidos, no dirigidos, ponderados y no ponderados.
