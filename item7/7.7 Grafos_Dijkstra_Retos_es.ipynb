## 📌 Ideas fundamentales

Dijkstra calcula las distancias mínimas desde un origen a los demás vértices de un grafo ponderado, siempre que los pesos sean **no negativos**.

La implementación básica de esta práctica sigue las tres estructuras del PDF:

- `distance`: distancia mínima provisional.
- `previous`: nodo anterior del mejor camino conocido.
- `visited`: indica si la distancia ya se considera definitiva.

En cada iteración se selecciona el nodo no visitado de menor distancia, se marca como visitado y se evalúan sus vecinos mediante la operación de **relajación**.

Con listas/diccionarios y búsqueda lineal del mínimo, la implementación básica tiene coste `O(V² + E)`.


## 🔁 ¿Dijkstra, BFS o DFS?

Antes de resolver un problema, identifica qué se está optimizando:

| Objetivo | Algoritmo adecuado |
|---|---|
| Visitar / comprobar accesibilidad | BFS o DFS |
| Menor número de aristas en grafo no ponderado | BFS |
| Menor suma de pesos no negativos | Dijkstra |

DFS puede encontrar *un* camino, pero no garantiza que sea el de menor número de aristas ni el de menor peso.


## Clase base

Partimos de la clase `Graph` basada en diccionarios y listas de adyacencia. Se incluyen también BFS y DFS de la práctica anterior para los ejercicios que requieren comparar o seleccionar algoritmos.


```python
from collections import deque
import math

class AdjacentVertex:
    def __init__(self, vertex, weight=1):
        self.vertex = vertex
        self.weight = weight

class Graph:
    def __init__(self, vertices: list, directed: bool):
        self._vertices = {v: [] for v in vertices}
        self._directed = directed

    def add_vertex(self, vertex):
        if vertex not in self._vertices:
            self._vertices[vertex] = []

    def add_edge(self, start, end, weight=1):
        if start not in self._vertices or end not in self._vertices:
            return
        self._vertices[start].append(AdjacentVertex(end, weight))
        if not self._directed:
            self._vertices[end].append(AdjacentVertex(start, weight))

    def get_adjacents(self, vertex):
        if vertex not in self._vertices:
            return []
        return [adj.vertex for adj in self._vertices[vertex]]

    def bfs(self, origin):
        if origin not in self._vertices:
            return []
        visited = {origin}
        queue = deque([origin])
        result = []
        while queue:
            u = queue.popleft()
            result.append(u)
            for v in self.get_adjacents(u):
                if v not in visited:
                    visited.add(v)
                    queue.append(v)
        return result

    def dfs(self, origin):
        if origin not in self._vertices:
            return []
        visited, result = set(), []

        def visit(u):
            visited.add(u)
            result.append(u)
            for v in self.get_adjacents(u):
                if v not in visited:
                    visit(v)

        visit(origin)
        return result
```
