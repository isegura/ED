> ## 📌 Lab Rules
>
> - We will work with the `Graph` class and the internal attribute `self._vertices` from the pseudocode of the topic.
> - `self._vertices[v]` stores the **outgoing** adjacent vertices from `v`.
> - Each connection is represented by an `AdjacentVertex(vertex, weight)` object; **it is not a tuple**.
> - In unweighted graphs, `weight = 1` is used by default.
> - In an undirected graph, an edge `u-v` must be stored in both directions.
> - When a query method receives a non-existent vertex, it will return a neutral result (`[]`, `0`, or `False`, depending on the method) instead of modifying the graph.


## Representation Used

The topic represents the graph using a dictionary:

```python
{
    'A': [AdjacentVertex('B', 5), AdjacentVertex('C', 1)],
    'B': [AdjacentVertex('A', 5)],
    'C': [AdjacentVertex('A', 1)]
}
```

The key is the origin vertex. The associated list contains its destinations and weights. This representation allows working naturally with directed, undirected, weighted, and unweighted graphs.
