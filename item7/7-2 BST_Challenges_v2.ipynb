# Lab Practice: Binary Search Tree (BST) ADT

This lab practice starts from **BinaryNode** and **BinaryTree**. The specific operations of a BST will be implemented in **BinarySearchTree**, which inherits from `BinaryTree`.

A **Binary Search Tree (BST)** maintains the following property:

- All elements in the left subtree are **smaller** than the node's element.
- All elements in the right subtree are **greater**.
- The property applies recursively to all subtrees.
- In this practice, **duplicate elements are not allowed**.

First, we will implement the fundamental operations covered in the topic (`search`, `insert`, and `remove`), and then we will solve additional problems by taking advantage of the ordering property of the BST.

## 📚 Final Summary: `BinarySearchTree` implementation

The following cell gathers only the **basic BST implementation** that we want to keep as a reference at the end of the lab practice.

It includes:

- `BinaryNode` and `BinaryTree` as the base structure
- `BinarySearchTree(BinaryTree)` as a derived class for BST operations
- `search` — iterative and recursive versions
- `insert` — iterative and recursive versions
- `remove`
- `predecessor`
- `successor`

The rest of the methods developed in the exercises **are not included in this final summary**, so that this cell remains focused on the fundamental operations of the Binary Search Tree ADT.


````python
class BinaryNode:
    def __init__(self, e):
        self.element = e
        self.left = None
        self.right = None


class BinaryTree:
    def __init__(self):
        self.root = None

    # Depth-first traversals
    def preorder(self, node):
        if node is None:
            return []
        return [node.element] + self.preorder(node.left) + self.preorder(node.right)

    def inorder(self, node):
        if node is None:
            return []
        return self.inorder(node.left) + [node.element] + self.inorder(node.right)

    def postorder(self, node):
        if node is None:
            return []
        return self.postorder(node.left) + self.postorder(node.right) + [node.element]

    # Number of nodes and height
    def size(self, node):
        if node is None:
            return 0
        return 1 + self.size(node.left) + self.size(node.right)

    def height(self, node):
        if node is None:
            return -1
        return 1 + max(self.height(node.left), self.height(node.right))


class BinarySearchTree(BinaryTree):
    # --------------------------------------------------------
    # SEARCH
    # --------------------------------------------------------

    def searchIte(self, x):
        current = self.root

        while current is not None:
            if current.element == x:
                return True

            if x < current.element:
                current = current.left
            else:
                current = current.right

        return False

    def search(self, x):
        return self._searchNode(self.root, x)

    def _searchNode(self, node, x):
        if node is None:
            return False

        if node.element == x:
            return True

        if x < node.element:
            return self._searchNode(node.left, x)

        return self._searchNode(node.right, x)

    # --------------------------------------------------------
    # INSERT
    # --------------------------------------------------------

    def insertIte(self, x):
        if self.root is None:
            self.root = BinaryNode(x)
            return

        node = self.root

        while node is not None:
            if x < node.element:
                if node.left is None:
                    node.left = BinaryNode(x)
                    return
                node = node.left

            elif x > node.element:
                if node.right is None:
                    node.right = BinaryNode(x)
                    return
                node = node.right

            else:
                # Duplicates are not allowed
                return

    def insert(self, x):
        self.root = self._insertNode(self.root, x)

    def _insertNode(self, node, x):
        if node is None:
            return BinaryNode(x)

        if node.element == x:
            return node

        if x < node.element:
            node.left = self._insertNode(node.left, x)
        else:
            node.right = self._insertNode(node.right, x)

        return node

    # --------------------------------------------------------
    # PREDECESSOR
    # --------------------------------------------------------

    def predecessor(self, node):
        if node is None or node.left is None:
            return None

        current = node.left

        while current.right is not None:
            current = current.right

        return current

    # --------------------------------------------------------
    # SUCCESSOR
    # --------------------------------------------------------

    def successor(self, node):
        if node is None or node.right is None:
            return None

        current = node.right

        while current.left is not None:
            current = current.left

        return current

    # --------------------------------------------------------
    # REMOVE
    # --------------------------------------------------------

    def getSuccessor(self, node):
        current = node

        while current.left is not None:
            current = current.left

        return current

    def remove(self, x):
        self.root = self._removeNode(self.root, x)

    def _removeNode(self, node, x):
        if node is None:
            return node

        if x < node.element:
            node.left = self._removeNode(node.left, x)

        elif x > node.element:
            node.right = self._removeNode(node.right, x)

        else:
            # Case 1 and case 2: zero or one child
            if node.left is None:
                return node.right

            if node.right is None:
                return node.left

            # Case 3: two children.
            # Substitute with the successor:
            # minimum of the right subtree.
            successor = self.getSuccessor(node.right)
            node.element = successor.element

            node.right = self._removeNode(
                node.right,
                successor.element
            )

        return node
````
