**Conventions:** in a BST, elements in the left subtree are smaller and elements in the right subtree are greater than the root element. **Duplicates are not allowed.** Height is measured in **edges**: empty tree `-1`, leaf `0`. Tests use numeric values.


## Challenge 1 · Search for an element

**Goal:** implement `searchIte(x)` (iterative) and `search(x)` (recursive, using `_searchNode(node, x)`). Both return `True` if the element exists and `False` otherwise.

**Before you start:** if the target value is smaller or greater than the current node's value, why can you discard one of the two subtrees?

The next cell contains the classes and **all the methods you will complete throughout this practice**. Run it again whenever you modify a `TODO`.


````python
class BinaryNode:
    def __init__(self, element):
        self.element = element
        self.left = None
        self.right = None


class BinaryTree:
    def __init__(self):
        self.root = None

    def preorder(self):
        return self._preorder(self.root)

    def _preorder(self, node):
        if node is None:
            return []
        return [node.element] + self._preorder(node.left) + self._preorder(node.right)

    def inorder(self):
        return self._inorder(self.root)

    def _inorder(self, node):
        if node is None:
            return []
        return self._inorder(node.left) + [node.element] + self._inorder(node.right)

    def postorder(self):
        return self._postorder(self.root)

    def _postorder(self, node):
        if node is None:
            return []
        return self._postorder(node.left) + self._postorder(node.right) + [node.element]

    def size(self):
        return self._size(self.root)

    def _size(self, node):
        if node is None:
            return 0
        return 1 + self._size(node.left) + self._size(node.right)

    def height(self):
        return self._height(self.root)

    def _height(self, node):
        if node is None:
            return -1
        return 1 + max(self._height(node.left), self._height(node.right))

    def print_tree(self):
        if self.root is None:
            print('(empty tree)')
            return
        print(self.root.element)

        def show(node, prefix):
            children = []
            if node.left is not None:
                children.append(('L', node.left))
            if node.right is not None:
                children.append(('R', node.right))
            for index, (side, child) in enumerate(children):
                last = index == len(children) - 1
                print(prefix + ('└── ' if last else '├── ') + side + ': ' + str(child.element))
                show(child, prefix + ('    ' if last else '│   '))

        show(self.root, '')


class BinarySearchTree(BinaryTree):
    # Challenge 1: search
    def searchIte(self, x):
        node = self.root
        while node is not None:
            if x == node.element:
                return True
            if x < node.element:
                node = node.left
            else:
                node = node.right
        return False

    def search(self, x):
        return self._searchNode(self.root, x)

    def _searchNode(self, node, x):
        if node is None:
            return False
        if x == node.element:
            return True
        if x < node.element:
            return self._searchNode(node.left, x)
        return self._searchNode(node.right, x)

    # Challenge 2: insertion
    def insertIte(self, x):
        new_node = BinaryNode(x)
        if self.root is None:
            self.root = new_node
            return

        parent = None
        node = self.root
        while node is not None:
            parent = node
            if x == node.element:
                return
            if x < node.element:
                node = node.left
            else:
                node = node.right

        if x < parent.element:
            parent.left = new_node
        else:
            parent.right = new_node

    def insert(self, x):
        self.root = self._insertNode(self.root, x)

    def _insertNode(self, node, x):
        if node is None:
            return BinaryNode(x)
        if x < node.element:
            node.left = self._insertNode(node.left, x)
        elif x > node.element:
            node.right = self._insertNode(node.right, x)
        return node

    # Challenge 4: predecessor from the left subtree
    def predecessor(self, node):
        if node is None or node.left is None:
            return None
        current = node.left
        while current.right is not None:
            current = current.right
        return current

    # Challenge 5: successor from the right subtree
    def successor(self, node):
        if node is None or node.right is None:
            return None
        current = node.right
        while current.left is not None:
            current = current.left
        return current



    def remove(self, x):
        self.root = self._removeNode(self.root, x)

    def _removeNode(self, node, x):
        if node is None:
            return None

        if x < node.element:
            node.left = self._removeNode(node.left, x)
        elif x > node.element:
            node.right = self._removeNode(node.right, x)
        else:
            # Case 1: leaf node
            if node.left is None and node.right is None:
                return None

            # Case 2: node with exactly one child
            if node.left is None:
                return node.right
            if node.right is None:
                return node.left

            # Case 3: node with two children
            next_node = self.successor(node)
            node.element = next_node.element
            node.right = self._removeNode(node.right, next_node.element)

        return node

    # Challenge 7: distance between predecessor and successor
    def distPredSuc(self, node):
        # TODO
        pass

    # Challenge 8: iterative minimum and maximum
    def minimum(self):
        # TODO
        pass

    def maximum(self):
        # TODO
        pass

    # Challenge 9: recursive minimum and maximum
    def minimumRec(self):
        # TODO
        pass

    def maximumRec(self):
        # TODO
        pass

    # Challenge 10: values within a range
    def getCount(self, start, end):
        # TODO
        pass

    # Challenge 11: validation
    def isBST(self):
        # TODO
        pass

    # Challenge 12: closest value
    def closest(self, value):
        # TODO
        pass

    # Challenge 13: lowest common ancestor
    def lowestCommonAncestor(self, a, b):
        # TODO
        pass
````
