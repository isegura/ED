## Reto 2 · Construir árboles de prueba

Para las pruebas utilizaremos este árbol binario de ejemplo (**no es un árbol binario de búsqueda**):

```text
          8
        /   \
       3     10
      / \      \
     1   6      14
        / \    /
       4   7  13
```

**Construcción paso a paso:** primero creamos los nodos; después, conectamos explícitamente sus hijos izquierdo (`left`) y derecho (`right`). Finalmente, creamos el árbol indicando cuál es la raíz.

La función `example_tree()` permite obtener un **árbol nuevo** cada vez que una prueba lo necesite. **Ejecuta esta celda después de volver a ejecutar la definición de las clases.**
El método `print_tree()` está **resuelto** en `BinaryTree` y permite visualizar los enlaces entre nodos sin necesitar completar ningún reto. Las etiquetas `L` y `R` indican hijo izquierdo y derecho, respectivamente.


````python
# Árbol de referencia: construcción paso a paso.
def example_tree():
    # 1. Crear los nodos (todavía no están conectados).
    n8 = BinaryNode(8)
    n3 = BinaryNode(3)
    n10 = BinaryNode(10)
    n1 = BinaryNode(1)
    n6 = BinaryNode(6)
    n14 = BinaryNode(14)
    n4 = BinaryNode(4)
    n7 = BinaryNode(7)
    n13 = BinaryNode(13)

    # 2. Conectar los nodos según el dibujo.
    n8.left = n3
    n8.right = n10

    n3.left = n1
    n3.right = n6
    n10.right = n14

    n6.left = n4
    n6.right = n7
    n14.left = n13

    # 3. Crear el árbol a partir de su raíz.
    return BinaryTree(n8)


def empty_tree():
    return BinaryTree()


def single_tree(value=5):
    return BinaryTree(BinaryNode(value))


tree = example_tree()
assert tree.root.elem == 8
assert tree.root.left.right.elem == 6
assert empty_tree().root is None
print("Árboles de prueba construidos correctamente.")
# Visualizar el árbol antes de comenzar los retos.
tree.print_tree()
````
