# Simple Practice: B-Tree ADT

In this practice session, we will work with the basic concepts of a **B-Tree**.

A B-Tree is a self-balancing search tree in which:

- each node can store **multiple keys**;
- a node can have **more than two children**;
- keys within a node are ordered and separate the range of its subtrees;
- all leaves are at the same level.

This practice is designed to be **highly guided**. The coding portion focuses on the tree structure, `search`, and `insert`. The `remove` operation, which involves more rebalancing cases, will be addressed using **graphical exercises**, without requiring full code implementation.


## 📌 Used Convention

We will mainly work with **order 4 B-Trees**:

- maximum **4 children** per node;
- maximum **3 keys** per node;
- except for the root, nodes must retain at least **1 key**.

We will graphically represent a node by enclosing its keys in square brackets:

```text
[10 | 20]
```

and its children underneath:

```text
          [10 | 20]
        /     |      \
      [5]   [15]   [25 | 30]
```
