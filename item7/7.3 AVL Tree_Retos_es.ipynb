# Práctica gráfica: Árboles AVL

En esta práctica **todos los ejercicios se resuelven gráficamente**. No es necesario programar ninguna rotación.

Un árbol AVL es un **árbol binario de búsqueda auto-balanceado**: después de cada inserción o eliminación debemos comprobar el equilibrio desde el nodo modificado hacia la raíz.

### Convención de trabajo

En cada paso:

1. Inserta o elimina como en un BST.
2. Recorre el camino **de abajo hacia arriba**.
3. Localiza el **primer nodo desequilibrado**.
4. Identifica el caso: **LL, RR, LR o RL**.
5. Aplica la rotación correspondiente.
6. Dibuja el árbol resultante antes de continuar con la siguiente operación.

Usaremos `BF = altura(derecho) - altura(izquierdo)`. Un nodo está equilibrado cuando su factor pertenece a `{-1, 0, 1}`.


## 🧭 Recordatorio de rotaciones

- **LL (Left-Left)** → rotación simple **a la derecha**.
- **RR (Right-Right)** → rotación simple **a la izquierda**.
- **LR (Left-Right)** → rotación **a la izquierda** sobre el hijo izquierdo y después **a la derecha** sobre el nodo desequilibrado.
- **RL (Right-Left)** → rotación **a la derecha** sobre el hijo derecho y después **a la izquierda** sobre el nodo desequilibrado.

En los retos no basta con indicar el nombre de la rotación: hay que **dibujar el árbol después de cada inserción que produzca un desequilibrio**.

# 🎯 Fin de la práctica

Al terminar deberías ser capaz de reconocer y resolver gráficamente:

- desequilibrios **LL** y **RR** mediante rotaciones simples;
- desequilibrios **LR** y **RL** mediante rotaciones dobles;
- transferencia de subárboles durante una rotación;
- secuencias de inserciones con varios rebalanceos;
- rebalanceo después de una eliminación.

Recuerda: el análisis se realiza **de abajo hacia arriba después de cada inserción o eliminación**.
