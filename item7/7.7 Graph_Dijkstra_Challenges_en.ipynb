## 📌 Key Concepts

Dijkstra calculates the shortest distances from a source vertex to all other vertices in a weighted graph, provided that all edge weights are **non-negative**.

The basic implementation in this lab follows the three structures presented in the topic PDF:

- `distance`: provisional minimum distance.
- `previous`: predecessor node in the best known path.
- `visited`: indicates whether the distance is already considered definitive.

In each iteration, the unvisited node with the smallest distance is selected, marked as visited, and its neighbors are evaluated using the **edge relaxation** operation.

With lists/dictionaries and a linear search for the minimum distance, this basic implementation has a time complexity of `O(V² + E)`.


## 🔁 Dijkstra, BFS, or DFS?

Before solving a problem, identify what is being optimized:

| Objective | Appropriate Algorithm |
|---|---|
| Visit / check reachability | BFS or DFS |
| Minimum number of edges in an unweighted graph | BFS |
| Minimum sum of non-negative weights | Dijkstra |

DFS can find *a* path, but it does not guarantee the path with the fewest edges nor the one with the smallest total weight.


## Base Class

We start from the `Graph` class based on dictionaries and adjacency lists. BFS and DFS from the previous lab are also included for exercises that require comparing or selecting algorithms.


```python
from collections import deque
import math

class AdjacentVertex:
    def __init__(self, vertex, weight=1):
        self.vertex = vertex
        self.weight = weight

class Graph:
    def __init__(self, vertices: list, directed: bool):
        self._vertices = {v: [] for v in vertices}
        self._directed = directed

    def add_vertex(self, vertex):
        if vertex not in self._vertices:
            self._vertices[vertex] = []

    def add_edge(self, start, end, weight=1):
        if start not in self._vertices or end not in self._vertices:
            return
        self._vertices[start].append(AdjacentVertex(end, weight))
        if not self._directed:
            self._vertices[end].append(AdjacentVertex(start, weight))

    def get_adjacents(self, vertex):
        if vertex not in self._vertices:
            return []
        return [adj.vertex for adj in self._vertices[vertex]]

    def bfs(self, origin):
        if origin not in self._vertices:
            return []
        visited = {origin}
        queue = deque([origin])
        result = []
        while queue:
            u = queue.popleft()
            result.append(u)
            for v in self.get_adjacents(u):
                if v not in visited:
                    visited.add(v)
                    queue.append(v)
        return result

    def dfs(self, origin):
        if origin not in self._vertices:
            return []
        visited, result = set(), []

        def visit(u):
            visited.add(u)
            result.append(u)
            for v in self.get_adjacents(u):
                if v not in visited:
                    visit(v)

        visit(origin)
        return result
```
