## Challenge 2 · Build test trees

We will use this example binary tree for testing (**it is not a binary search tree**):

```text
          8
        /   \
       3     10
      / \      \
     1   6      14
        / \    /
       4   7  13
```

**Step by step:** First create the nodes; then explicitly connect their left (`left`) and right (`right`) children. Finally, create the tree by specifying its root.

The `example_tree()` function creates a **fresh tree** whenever a test needs one. **Run this cell after rerunning the class definitions.**
The `print_tree()` method is **already implemented** in `BinaryTree`; it displays node links without completing any challenge. Labels `L` and `R` indicate left and right children.


````python
# Reference tree: step-by-step construction.
def example_tree():
    # 1. Create nodes (not connected yet).
    n8 = BinaryNode(8)
    n3 = BinaryNode(3)
    n10 = BinaryNode(10)
    n1 = BinaryNode(1)
    n6 = BinaryNode(6)
    n14 = BinaryNode(14)
    n4 = BinaryNode(4)
    n7 = BinaryNode(7)
    n13 = BinaryNode(13)

    # 2. Connect nodes according to the diagram.
    n8.left = n3
    n8.right = n10

    n3.left = n1
    n3.right = n6
    n10.right = n14

    n6.left = n4
    n6.right = n7
    n14.left = n13

    # 3. Create the tree from its root.
    return BinaryTree(n8)


def empty_tree():
    return BinaryTree()


def single_tree(value=5):
    return BinaryTree(BinaryNode(value))


tree = example_tree()
assert tree.root.elem == 8
assert tree.root.left.right.elem == 6
assert empty_tree().root is None
print("Test trees built successfully.")
# Visualize the tree before starting the challenges.
tree.print_tree()
````
