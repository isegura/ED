## 📌 Key Concepts

A traversal must avoid processing the same vertex multiple times, because in a graph a node can be reached via different paths.

- **BFS (Breadth-First Search)** explores by proximity levels and uses a **FIFO queue**.
- **DFS (Depth-First Search)** goes deep along a branch and then backtracks. In this topic, it is developed **recursively**.
- With adjacency lists, both traversals have a time complexity of **O(V + E)**.
- The exact traversal order depends on the order of the vertices in the adjacency lists.


## Base Class

We start from the representation from the previous lab. `get_adjacents(vertex)` returns only the identifiers of the adjacent vertices, not their weights.


```python
from collections import deque

class AdjacentVertex:
    def __init__(self, vertex, weight=1):
        self.vertex = vertex
        self.weight = weight

class Graph:
    def __init__(self, vertices: list, directed: bool):
        self._vertices = {}
        for v in vertices:
            self._vertices[v] = []
        self._directed = directed

    def add_vertex(self, vertex):
        if vertex in self._vertices:
            return
        self._vertices[vertex] = []

    def add_edge(self, start, end, weight=1):
        if start not in self._vertices or end not in self._vertices:
            return
        self._vertices[start].append(AdjacentVertex(end, weight))
        if not self._directed:
            self._vertices[end].append(AdjacentVertex(start, weight))

    def get_adjacents(self, vertex):
         if vertex not in self._vertices:
            print(vertex, ' does not exist!')
            return []

         adjacents = []
         for adj in self._vertices[vertex]:
            adjacents.append(adj.vertex)

         return adjacents
```
