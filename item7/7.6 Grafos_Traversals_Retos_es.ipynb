## 📌 Ideas del tema

Un recorrido debe evitar procesar varias veces el mismo vértice, porque en un grafo un nodo puede alcanzarse por diferentes caminos.

- **BFS (Breadth-First Search)** explora por niveles de cercanía y utiliza una **cola FIFO**.
- **DFS (Depth-First Search)** profundiza por una rama y después retrocede (*backtracking*). En el tema se desarrolla de forma **recursiva**.
- Con listas de adyacencia, ambos recorridos tienen complejidad temporal **O(V + E)**.
- El orden concreto del recorrido depende del orden de los vértices en las listas de adyacencia.


## Clase base

Partimos de la representación de la práctica anterior. `get_adjacents(vertex)` devuelve solo los identificadores de los vértices adyacentes, no sus pesos.


```python
from collections import deque

class AdjacentVertex:
    def __init__(self, vertex, weight=1):
        self.vertex = vertex
        self.weight = weight

class Graph:
    def __init__(self, vertices: list, directed: bool):
        self._vertices = {}
        for v in vertices:
            self._vertices[v] = []
        self._directed = directed

    def add_vertex(self, vertex):
        if vertex in self._vertices:
            return
        self._vertices[vertex] = []

    def add_edge(self, start, end, weight=1):
        if start not in self._vertices or end not in self._vertices:
            return
        self._vertices[start].append(AdjacentVertex(end, weight))
        if not self._directed:
            self._vertices[end].append(AdjacentVertex(start, weight))

    def get_adjacents(self, vertex):
        if vertex not in self._vertices:
            return []
        return [adj.vertex for adj in self._vertices[vertex]]
```
