- **`search(key)`:** método público que inicia la búsqueda desde la raíz del árbol.
- **`_search_recursive(node, key)`:** método auxiliar que realiza la búsqueda recursiva a partir de un nodo.

**Valor devuelto:** Si encuentra la clave, devuelve una tupla `(node, i)`, donde `node` es el nodo que contiene la clave e `i` es su posición en `node.keys`. Si no la encuentra, devuelve `None`.

**Tiempo estimado:** 15 minutos · **Dificultad:** Fácil

4. En caso contrario, continúa la búsqueda recursivamente por `node.children[i]`.

**Recuerda:** Antes de acceder a `node.keys[i]`, comprueba que `i` es una posición válida.

# 🏆 Reto 8 · Fusionar nodos hermanos

El método debe:

1. **Seleccionar un hermano adyacente**, priorizando el izquierdo si existe.
2. **Fusionar las claves** de ambos nodos junto con la clave separadora del padre, manteniendo el orden.
3. Si son nodos internos, **combinar también sus hijos** y actualizar sus referencias `parent`.
4. **Actualizar el padre**, eliminando la clave separadora y la referencia al nodo que desaparece.
5. **Devolver el padre**, para comprobar posteriormente si también necesita reestructuración.

**Importante:** Este método realiza una única fusión. No comprueba si el padre queda deficitario ni propaga la reestructuración hacia arriba. Estas operaciones se abordarán en el siguiente reto.

🕒 **Tiempo estimado:** 25 minutos  
🟠 **Dificultad:** Intermedia-alta

### 💡 Pista

Utiliza `parent.children.index(node)` para localizar la posición del nodo entre sus hermanos.

Recuerda que:

- Si fusionas con el **hermano izquierdo**, la clave separadora del padre es `parent.keys[i - 1]`.
- Si fusionas con el **hermano derecho**, la clave separadora es `parent.keys[i]`.

Puedes utilizar los métodos `pop()`, `append()` y `extend()` de las listas de Python.

Esta celda comprueba el Reto 8, incluyendo fusiones con el hermano izquierdo y derecho, tanto de hojas como de nodos internos.

# 🏆 Reto 9 · Corregir el déficit de claves tras una eliminación

🎯 **Objetivo:** Implementa el método `_fix_underflow(node)` de la clase `BTree`, que permite restaurar las propiedades del árbol B cuando un nodo tiene menos claves que el mínimo permitido.

El método debe:

1. **Comprobar si el nodo tiene déficit de claves.** Recuerda que, excepto la raíz, cada nodo debe contener al menos `ceil(m / 2) - 1` claves.
2. **Intentar una redistribución:** si algún hermano puede prestar una clave, utilizar `_borrow_from_left()` o `_borrow_from_right()`, dando prioridad al hermano izquierdo.
3. **Realizar una fusión:** si ningún hermano puede prestar una clave, utilizar `_merge(node)`.
4. **Propagar la corrección hacia arriba:** después de una fusión, comprobar si el padre también ha quedado deficitario y, en ese caso, continuar el proceso.
5. **Actualizar la raíz:** si la raíz queda sin claves, sustituirla por su único hijo o establecer `self.root = None` si el árbol ha quedado vacío.

**Importante:** Este método no elimina ninguna clave. Su responsabilidad es corregir los posibles déficits que se hayan producido durante la eliminación.

🕒 **Tiempo estimado:** 25 minutos  
🔴 **Dificultad:** Alta

### 💡 Pista

Utiliza un bucle `while` para corregir los déficits que puedan propagarse hacia la raíz.

En cada iteración:

- Calcula el mínimo de claves permitido.
- Comprueba primero si el hermano izquierdo puede prestar una clave y, después, si puede hacerlo el derecho.
- Si ninguno puede prestar, fusiona los nodos y continúa comprobando el padre.

Recuerda que puedes utilizar los métodos implementados en los retos anteriores: `_borrow_from_left()`, `_borrow_from_right()` y `_merge()`.


Aquí tienes una celda de tests para el Reto 9, con árboles construidos manualmente para no depender de insert() ni de remove().

Se comprueban las redistribuciones, las fusiones, la propagación del déficit y la reducción de la altura del árbol.

````python
# Tests Reto 9: BTree._fix_underflow()

# Test 1: Redistribución desde el hermano izquierdo
tree = BTree(4)

root = BTreeNode(is_leaf=False)
root.keys = [20]
tree.root = root

left = BTreeNode()
left.keys = [5, 10]

node = BTreeNode()
node.keys = []  # Nodo deficitario

root.children = [left, node]
left.parent = root
node.parent = root

tree._fix_underflow(node)

assert root.keys == [10]
assert left.keys == [5]
assert node.keys == [20]
assert root.children == [left, node]


# Test 2: Redistribución desde el hermano derecho
tree = BTree(4)

root = BTreeNode(is_leaf=False)
root.keys = [20]
tree.root = root

node = BTreeNode()
node.keys = []  # Nodo deficitario

right = BTreeNode()
right.keys = [30, 40]

root.children = [node, right]
node.parent = root
right.parent = root

tree._fix_underflow(node)

assert root.keys == [30]
assert node.keys == [20]
assert right.keys == [40]


# Test 3: Prioridad del hermano izquierdo
tree = BTree(4)

root = BTreeNode(is_leaf=False)
root.keys = [20, 40]
tree.root = root

left = BTreeNode()
left.keys = [5, 10]

node = BTreeNode()
node.keys = []  # Nodo deficitario

right = BTreeNode()
right.keys = [50, 60]

root.children = [left, node, right]
for child in root.children:
    child.parent = root

tree._fix_underflow(node)

assert root.keys == [10, 40]
assert left.keys == [5]
assert node.keys == [20]
assert right.keys == [50, 60]


# Test 4: Fusión sin propagación del déficit
tree = BTree(4)

root = BTreeNode(is_leaf=False)
root.keys = [10, 20]
tree.root = root

left = BTreeNode()
left.keys = [5]

node = BTreeNode()
node.keys = []  # Nodo deficitario

right = BTreeNode()
right.keys = [25]

root.children = [left, node, right]
for child in root.children:
    child.parent = root

tree._fix_underflow(node)

assert root.keys == [20]
assert root.children == [left, right]
assert left.keys == [5, 10]
assert right.keys == [25]
assert tree.root is root


# Test 5: Fusión con reducción de la altura
tree = BTree(4)

root = BTreeNode(is_leaf=False)
root.keys = [20]
tree.root = root

node = BTreeNode()
node.keys = []  # Nodo deficitario

right = BTreeNode()
right.keys = [30]

root.children = [node, right]
node.parent = root
right.parent = root

tree._fix_underflow(node)

assert tree.root is node
assert tree.root.keys == [20, 30]
assert tree.root.is_leaf
assert tree.root.parent is None


# Test 6: Propagación del déficit hasta la raíz
tree = BTree(4)

root = BTreeNode(is_leaf=False)
root.keys = [30]
tree.root = root

left = BTreeNode(is_leaf=False)
left.keys = [10]

right = BTreeNode(is_leaf=False)
right.keys = [40]

root.children = [left, right]
left.parent = root
right.parent = root

a = BTreeNode()
a.keys = [5]

node = BTreeNode()
node.keys = []  # Nodo deficitario

b = BTreeNode()
b.keys = [35]

c = BTreeNode()
c.keys = [45]

left.children = [a, node]
right.children = [b, c]

for child in left.children:
    child.parent = left
for child in right.children:
    child.parent = right

tree._fix_underflow(node)

# Primera fusión: [5] + 10 + [] -> [5, 10]
# Segunda fusión: [40] + 30 + [10] -> [10, 30, 40]
# La raíz original desaparece.

assert tree.root is left
assert tree.root.keys == [30, 40]
assert tree.root.children == [a, b, c]
assert a.keys == [5, 10]
assert b.keys == [35]
assert c.keys == [45]
assert tree.root.parent is None

for child in tree.root.children:
    assert child.parent is tree.root


# Test 7: El árbol queda vacío
tree = BTree(4)

root = BTreeNode()
root.keys = []
tree.root = root

tree._fix_underflow(root)

assert tree.root is None


# Test 8: Nodo sin déficit (no debe modificarse)
tree = BTree(4)

root = BTreeNode(is_leaf=False)
root.keys = [20]
tree.root = root

left = BTreeNode()
left.keys = [10]

right = BTreeNode()
right.keys = [30]

root.children = [left, right]
left.parent = root
right.parent = root

tree._fix_underflow(left)

assert tree.root is root
assert root.keys == [20]
assert left.keys == [10]
assert right.keys == [30]
assert root.children == [left, right]

print("Reto 9: Todos los tests superados correctamente.")
````
