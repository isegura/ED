La práctica propone implementar el TAD Grafo usando un **diccionarios Python**, donde cada vértice es una clave del diccionario y su valor es una lista de objetos `AdjacentVertex`, cada uno con `vertex` y `weight`.

Por ejemplo, el siguiente grafo:

```text
          A
         / \
      5 /   \ 1
       /     \
      B       C
```

Se puede representa con el siguiente diccionario:


```python
{
    'A': [AdjacentVertex('B', 5), AdjacentVertex('C', 1)],
    'B': [AdjacentVertex('A', 5)],
    'C': [AdjacentVertex('A', 1)]
}
```
donde cada clave representa un vértice del grafo.
El valor asociado a cada clave es una lista de objetos AdjacentVertex. Cada objeto AdjacentVertex, tiene dos atributos: el vértice adyacente, y el peso de la arista.


Esta representación permite trabajar con grafos dirigidos, no dirigidos, ponderados y no ponderados.
