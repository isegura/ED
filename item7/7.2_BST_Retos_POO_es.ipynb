````python
class BinaryNode:
    def __init__(self, element):
        self.element = element
        self.left = None
        self.right = None


class BinaryTree:
    def __init__(self):
        self.root = None

    def preorder(self):
        return self._preorder(self.root)

    def _preorder(self, node):
        if node is None:
            return []
        return [node.element] + self._preorder(node.left) + self._preorder(node.right)

    def inorder(self):
        return self._inorder(self.root)

    def _inorder(self, node):
        if node is None:
            return []
        return self._inorder(node.left) + [node.element] + self._inorder(node.right)

    def postorder(self):
        return self._postorder(self.root)

    def _postorder(self, node):
        if node is None:
            return []
        return self._postorder(node.left) + self._postorder(node.right) + [node.element]

    def size(self):
        return self._size(self.root)

    def _size(self, node):
        if node is None:
            return 0
        return 1 + self._size(node.left) + self._size(node.right)

    def height(self):
        return self._height(self.root)

    def _height(self, node):
        if node is None:
            return -1
        return 1 + max(self._height(node.left), self._height(node.right))

    def print_tree(self):
        if self.root is None:
            print('(árbol vacío)')
            return
        print(self.root.element)

        def show(node, prefix):
            children = []
            if node.left is not None:
                children.append(('L', node.left))
            if node.right is not None:
                children.append(('R', node.right))
            for index, (side, child) in enumerate(children):
                last = index == len(children) - 1
                print(prefix + ('└── ' if last else '├── ') + side + ': ' + str(child.element))
                show(child, prefix + ('    ' if last else '│   '))

        show(self.root, '')


class BinarySearchTree(BinaryTree):
    # Reto 1: búsqueda
    def searchIte(self, x):
        # TODO
        pass

    def search(self, x):
        # TODO
        pass

    def _searchNode(self, node, x):
        # TODO
        pass

    # Reto 2: inserción
    def insertIte(self, x):
        # TODO
        pass

    def insert(self, x):
        # TODO
        pass

    def _insertNode(self, node, x):
        # TODO
        pass

    # Reto 4: predecesor del subárbol izquierdo
    def predecessor(self, node):
        # TODO
        pass

    # Reto 5: sucesor del subárbol derecho
    def successor(self, node):
        # TODO
        pass

    # Reto 6: borrado
    def getSuccessor(self, node):
        # TODO
        pass

    def remove(self, x):
        # TODO
        pass

    def _removeNode(self, node, x):
        # TODO
        pass

    # Reto 7: distancia entre predecesor y sucesor
    def distPredSuc(self, node):
        # TODO
        pass

    # Reto 8: mínimo y máximo iterativos
    def minimum(self):
        # TODO
        pass

    def maximum(self):
        # TODO
        pass

    # Reto 9: mínimo y máximo recursivos
    def minimumRec(self):
        # TODO
        pass

    def maximumRec(self):
        # TODO
        pass

    # Reto 10: valores en un intervalo
    def getCount(self, start, end):
        # TODO
        pass

    # Reto 11: validación
    def isBST(self):
        # TODO
        pass

    # Reto 12: valor más cercano
    def closest(self, value):
        # TODO
        pass

    # Reto 13: ancestro común más bajo
    def lowestCommonAncestor(self, a, b):
        # TODO
        pass
````
