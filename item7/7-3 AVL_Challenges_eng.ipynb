# Graphical Practice: AVL Trees

In this practice **all exercises are solved graphically**. It is not necessary to program any rotation.

An AVL tree is a **self-balancing binary search tree**: after each insertion or deletion we must check the balance from the modified node up to the root.

### Working Convention

In each step:

1. Insert or delete as in a BST.
2. Traverse the path **from bottom to top**.
3. Locate the **first unbalanced node**.
4. Identify the case: **LL, RR, LR, or RL**.
5. Apply the corresponding rotation.
6. Draw the resulting tree before continuing with the next operation.

We will use `BF = height(right) - height(left)`. A node is balanced when its factor belongs to `{-1, 0, 1}`.


## 🧭 Rotations Reminder

- **LL (Left-Left)** → simple rotation **to the right**.
- **RR (Right-Right)** → simple rotation **to the left**.
- **LR (Left-Right)** → rotation **to the left** on the left child and then **to the right** on the unbalanced node.
- **RL (Right-Left)** → rotation **to the right** on the right child and then **to the left** on the unbalanced node.

In the challenges it is not enough to state the name of the rotation: you must **draw the tree after each insertion that causes an imbalance**.

# 🎯 End of Practice

Upon finishing, you should be able to recognize and graphically solve:

- **LL** and **RR** imbalances using simple rotations;
- **LR** and **RL** imbalances using double rotations;
- subtree transfer during a rotation;

- rebalancing after a deletion.

Remember: analysis is performed **from bottom to top after each insertion or deletion**.
