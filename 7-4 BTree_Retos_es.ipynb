# Práctica sencilla: TAD Árbol B (B-Tree)

En esta práctica trabajaremos con los conceptos básicos de un **Árbol B**.

Un Árbol B es un árbol de búsqueda balanceado en el que:

- cada nodo puede almacenar **varias claves**;
- un nodo puede tener **más de dos hijos**;
- las claves de un nodo están ordenadas y separan los rangos de sus subárboles;
- todas las hojas están al mismo nivel.

La práctica está planteada de forma **muy guiada**. La parte de programación se concentra en la estructura del árbol, `search` e `insert`. La operación `remove`, que requiere más casos de reajuste, se trabajará mediante ejercicios **gráficos**, sin pedir su implementación completa.


## 📌 Convención utilizada

Trabajaremos principalmente con **Árboles B de orden 4**:

- máximo **4 hijos** por nodo;
- máximo **3 claves** por nodo;
- salvo la raíz, los nodos deben conservar al menos **1 clave**.

Representaremos gráficamente un nodo escribiendo sus claves entre corchetes:

```text
[10 | 20]
```

y sus hijos debajo:

```text
          [10 | 20]
        /     |      \
      [5]   [15]   [25 | 30]
```

## 📚 Resumen de la implementación

La siguiente celda reúne la parte programada en esta práctica:

- `BTreeNode`
- `BTree`
- `insert_sorted`
- `search`
- `search_leaf`
- `insert` completo con **split**, promoción de la mediana y propagación hacia arriba

`remove` sigue sin implementarse en código. Se trabaja gráficamente para poder razonar con claridad sobre los casos de préstamo, fusión, borrado en nodos internos y propagación del déficit hacia niveles superiores.


````python
class BTreeNode:
    def __init__(self, is_leaf=True):
        self.keys = []
        self.children = []
        self.parent = None
        self.is_leaf = is_leaf


def insert_sorted(keys, key):
    if key in keys:
        return

    i = 0
    while i < len(keys) and keys[i] < key:
        i += 1

    keys.insert(i, key)


class BTree:
    def __init__(self, m):
        self.root = None
        self.m = m

    def search(self, key):
        if self.root is None:
            return None
        return self._search_recursive(self.root, key)

    def _search_recursive(self, node, key):
        i = 0

        while i < len(node.keys) and key > node.keys[i]:
            i += 1

        if i < len(node.keys) and node.keys[i] == key:
            return node, i

        if node.is_leaf:
            return None

        return self._search_recursive(node.children[i], key)

    def search_leaf(self, key):
        if self.root is None:
            return None

        node = self.root

        while not node.is_leaf:
            i = 0
            while i < len(node.keys) and key > node.keys[i]:
                i += 1
            node = node.children[i]

        return node

    def insert(self, key):
        if self.root is None:
            self.root = BTreeNode(is_leaf=True)
            self.root.keys.append(key)
            return True

        if self.search(key) is not None:
            return False

        node = self.search_leaf(key)
        insert_sorted(node.keys, key)

        while len(node.keys) > self.m - 1:
            median_idx = len(node.keys) // 2
            median = node.keys[median_idx]

            left = BTreeNode(is_leaf=node.is_leaf)
            right = BTreeNode(is_leaf=node.is_leaf)

            left.keys = node.keys[:median_idx]
            right.keys = node.keys[median_idx + 1:]

            if not node.is_leaf:
                left.children = node.children[:median_idx + 1]
                right.children = node.children[median_idx + 1:]

                for child in left.children:
                    child.parent = left
                for child in right.children:
                    child.parent = right

            if node is self.root:
                new_root = BTreeNode(is_leaf=False)
                new_root.keys = [median]
                new_root.children = [left, right]
                left.parent = new_root
                right.parent = new_root
                self.root = new_root
                return True

            parent = node.parent
            left.parent = parent
            right.parent = parent

            pos_node = parent.children.index(node)
            parent.children[pos_node] = left
            parent.children.insert(pos_node + 1, right)

            insert_sorted(parent.keys, median)
            node = parent

        return True
````
