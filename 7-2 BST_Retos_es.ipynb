# Práctica: TAD Árbol Binario de Búsqueda (BST)

Esta práctica continúa directamente la implementación de **BinaryNode** y **BinaryTree** de la práctica anterior.

Un **Binary Search Tree (BST)** mantiene la propiedad:

- Todos los elementos del subárbol izquierdo son **menores** que el elemento del nodo.
- Todos los elementos del subárbol derecho son **mayores**.
- La propiedad se aplica recursivamente a todos los subárboles.
- En esta práctica **no se permiten elementos duplicados**.

Primero implementaremos las operaciones fundamentales del tema (`search`, `insert` y `remove`) y después resolveremos problemas adicionales aprovechando la propiedad de orden del BST.

## 📚 Resumen final: implementación de `BinarySearchTree`

La siguiente celda reúne únicamente la **implementación básica del BST** que queremos conservar como referencia al finalizar la práctica.

Incluye:

- `BinaryNode` y `BinaryTree`
- `search` — versión iterativa y recursiva
- `insert` — versión iterativa y recursiva
- `remove`
- `predecessor`
- `successor`

El resto de métodos desarrollados en los ejercicios **no se incluyen en este resumen final**, para que esta celda quede centrada en las operaciones fundamentales del TAD Binary Search Tree.


````python
class BinaryNode:
    def __init__(self, e):
        self.element = e
        self.left = None
        self.right = None


class BinaryTree:
    def __init__(self):
        self.root = None

    # --------------------------------------------------------
    # SEARCH
    # --------------------------------------------------------

    def searchIte(self, x):
        current = self.root

        while current is not None:
            if current.element == x:
                return True

            if x < current.element:
                current = current.left
            else:
                current = current.right

        return False

    def search(self, x):
        return self._searchNode(self.root, x)

    def _searchNode(self, node, x):
        if node is None:
            return False

        if node.element == x:
            return True

        if x < node.element:
            return self._searchNode(node.left, x)

        return self._searchNode(node.right, x)

    # --------------------------------------------------------
    # INSERT
    # --------------------------------------------------------

    def insertIte(self, x):
        if self.root is None:
            self.root = BinaryNode(x)
            return

        node = self.root

        while node is not None:
            if x < node.element:
                if node.left is None:
                    node.left = BinaryNode(x)
                    return
                node = node.left

            elif x > node.element:
                if node.right is None:
                    node.right = BinaryNode(x)
                    return
                node = node.right

            else:
                # No se permiten duplicados
                return

    def insert(self, x):
        self.root = self._insertNode(self.root, x)

    def _insertNode(self, node, x):
        if node is None:
            return BinaryNode(x)

        if node.element == x:
            return node

        if x < node.element:
            node.left = self._insertNode(node.left, x)
        else:
            node.right = self._insertNode(node.right, x)

        return node

    # --------------------------------------------------------
    # PREDECESSOR
    # --------------------------------------------------------

    def predecessor(self, node):
        if node is None or node.left is None:
            return None

        current = node.left

        while current.right is not None:
            current = current.right

        return current

    # --------------------------------------------------------
    # SUCCESSOR
    # --------------------------------------------------------

    def successor(self, node):
        if node is None or node.right is None:
            return None

        current = node.right

        while current.left is not None:
            current = current.left

        return current

    # --------------------------------------------------------
    # REMOVE
    # --------------------------------------------------------

    def getSuccessor(self, node):
        current = node

        while current.left is not None:
            current = current.left

        return current

    def remove(self, x):
        self.root = self._removeNode(self.root, x)

    def _removeNode(self, node, x):
        if node is None:
            return node

        if x < node.element:
            node.left = self._removeNode(node.left, x)

        elif x > node.element:
            node.right = self._removeNode(node.right, x)

        else:
            # Caso 1 y caso 2: cero o un hijo
            if node.left is None:
                return node.right

            if node.right is None:
                return node.left

            # Caso 3: dos hijos.
            # Sustituimos por el sucesor:
            # mínimo del subárbol derecho.
            successor = self.getSuccessor(node.right)
            node.element = successor.element

            node.right = self._removeNode(
                node.right,
                successor.element
            )

        return node
````
