# 🧱 Base code

The `SNode` class is fully implemented.

The `SList` class also provides the constructor and the basic operations for checking the size and adding elements at the beginning and end.

**Do not modify the code that is already implemented.**

````python
class SNode:
    """Node for a singly linked list."""

    def __init__(self, e, next=None):
        self.elem = e
        self.next = next


class SList:
    """Implementation of a singly linked list with a reference to the first node (head)."""

    def __init__(self):
        # By convention, the _ prefix indicates that these attributes are internal
        # to the class and should not be accessed directly from outside.
        self._head = None
        self._size = 0

    def __len__(self):
        return self._size

    def isEmpty(self):
        return len(self) == 0

    def __str__(self):
        result = ""
        current = self._head

        while current:
            result += str(current.elem)

            if current.next is not None:
                result += ","

            current = current.next

        return result

    def addFirst(self, e):
        newNode = SNode(e)
        newNode.next = self._head
        self._head = newNode
        self._size += 1

    def addLast(self, e):
        newNode = SNode(e)

        if self.isEmpty():
            self._head = newNode
        else:
            lastNode = self._head

            while lastNode.next:
                lastNode = lastNode.next

            lastNode.next = newNode

        self._size += 1

    def removeFirst(self):
        if self.isEmpty():
            return None

        result = self._head.elem
        self._head = self._head.next
        self._size -= 1

        return result

    def removeLast(self):
        if self.isEmpty():
            return None

        if len(self) == 1:
            return self.removeFirst()

        previous = None
        current = self._head

        while current.next:
            previous = current
            current = current.next

        previous.next = None
        self._size -= 1

        return current.elem

    def getAt(self, index):
        if index < 0 or index >= len(self):
            return None

        current = self._head

        for _ in range(index):
            current = current.next

        return current.elem

    def index(self, e):
        current = self._head
        position = 0

        while current:
            if current.elem == e:
                return position

            current = current.next
            position += 1

        return -1

    def insertAt(self, index, e):
        if index < 0 or index > len(self):
            return

        if index == 0:
            self.addFirst(e)
            return

        if index == len(self):
            self.addLast(e)
            return

        previous = self._head

        for _ in range(index - 1):
            previous = previous.next

        newNode = SNode(e)
        newNode.next = previous.next
        previous.next = newNode
        self._size += 1

    def removeAt(self, index):
        if index < 0 or index >= len(self):
            return None

        if index == 0:
            return self.removeFirst()

        previous = self._head

        for _ in range(index - 1):
            previous = previous.next

        current = previous.next
        previous.next = current.next
        self._size -= 1

        return current.elem
````
