# 🧱 Base code: `SNode` and `SList`

The `SList` class stores:

- `_head`: a reference to the first node.
- `_size`: the number of stored elements.

Each `SNode` contains an element (`elem`) and a reference to the next node (`next`).

**Run this cell before starting the challenges and do not modify it.**

````python
class SNode:
    """Node for a singly linked list."""

    def __init__(self, e, next=None):
        self.elem = e
        self.next = next


class SList:
    """Implementation of a singly linked list with a reference to the first node (head)."""

    def __init__(self):
        self._head = None
        self._size = 0

    def __len__(self):
        return self._size

    def isEmpty(self):
        return len(self) == 0

    def __str__(self):
        result = ''
        nodeIt = self._head
        while nodeIt:
            result += ',' + str(nodeIt.elem)
            nodeIt = nodeIt.next
        if len(result) > 0:
            result = result[1:]
        return result

    def addFirst(self, e):
        newNode = SNode(e)
        newNode.next = self._head
        self._head = newNode
        self._size += 1

    def addLast(self, e):
        newNode = SNode(e)
        if self.isEmpty():
            self._head = newNode
        else:
            lastNode = self._head
            while lastNode.next:
                lastNode = lastNode.next
            lastNode.next = newNode
        self._size += 1

    def removeFirst(self):
        result = None
        if self.isEmpty():
            print('Error: list is empty!')
        else:
            result = self._head.elem
            self._head = self._head.next
            self._size -= 1
        return result

    def removeLast(self):
        result = None
        if self.isEmpty():
            print('Error: list is empty!')
        elif len(self) == 1:
            result = self.removeFirst()
        else:
            penult = None
            lastNode = self._head
            while lastNode.next:
                penult = lastNode
                lastNode = lastNode.next
            result = lastNode.elem
            penult.next = None
            self._size -= 1
        return result

    def getAt(self, index):
        result = None
        if index not in range(0, len(self)):
            print(index, 'Error getAt: index out of range')
        else:
            nodeIt = self._head
            i = 0
            while nodeIt and i < index:
                nodeIt = nodeIt.next
                i += 1
            result = nodeIt.elem
        return result

    def index(self, e):
        nodeIt = self._head
        idx = 0
        while nodeIt:
            if nodeIt.elem == e:
                return idx
            nodeIt = nodeIt.next
            idx += 1
        return -1

    def insertAt(self, index, e):
        if index not in range(0, len(self) + 1):
            print(index, 'Error insertAt: index out of range')
        elif index == 0:
            self.addFirst(e)
        elif index == len(self):
            self.addLast(e)
        else:
            previous = self._head
            for _ in range(index - 1):
                previous = previous.next
            newNode = SNode(e)
            newNode.next = previous.next
            previous.next = newNode
            self._size += 1

    def removeAt(self, index):
        result = None
        if index not in range(len(self)):
            print(index, 'Error removeAt: index out of range')
        elif index == 0:
            result = self.removeFirst()
        elif index == len(self) - 1:
            result = self.removeLast()
        else:
            previous = self._head
            for _ in range(index - 1):
                previous = previous.next
            result = previous.next.elem
            previous.next = previous.next.next
            self._size -= 1
        return result
````
