# 🧱 Base code: `Queue` class

The following class implements a FIFO queue using a Python list internally.

**Run this cell before starting the challenges and do not modify it.**

````python
class Queue:
    """Implementation of a FIFO queue using a native Python list."""

    def __init__(self):
        """Initializes an empty queue."""
        self.items = []

    def __len__(self):
        """Returns the number of elements in the queue."""
        return len(self.items)

    def isEmpty(self):
        """Returns True if the queue is empty, False otherwise."""
        return len(self) == 0

    def __str__(self):
        """Returns a string representation of the queue elements."""
        return str(self.items)

    def enqueue(self, e):
        """Adds element e to the rear (tail) of the queue."""
        self.items.append(e)

    def dequeue(self):
        """Removes and returns the first element (head) of the queue."""
        if self.isEmpty():
            print("Error: The queue is empty")
            return None
        return self.items.pop(0)

    def front(self):
        """Returns the first element without removing it."""
        if self.isEmpty():
            print("Error: The queue is empty")
            return None
        return self.items[0]
````
