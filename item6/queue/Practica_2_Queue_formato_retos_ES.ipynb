# 🧱 Código base: clase `Queue`

La siguiente clase implementa una cola FIFO utilizando internamente una lista de Python.

**Ejecuta esta celda antes de comenzar los retos y no la modifiques.**

````python
class Queue:
    """Implementación de una Cola FIFO usando una lista nativa de Python."""

    def __init__(self):
        """Inicializa una cola vacía."""
        self.items = []

    def __len__(self):
        """Retorna la cantidad de elementos en la cola."""
        return len(self.items)

    def isEmpty(self):
        """Retorna True si la cola está vacía, False en caso contrario."""
        return len(self) == 0

    def __str__(self):
        """Retorna una representación en texto de los elementos."""
        return str(self.items)

    def enqueue(self, e):
        """Agrega el elemento e al final (tail) de la cola."""
        self.items.append(e)

    def dequeue(self):
        """Remueve y retorna el primer elemento (head) de la cola."""
        if self.isEmpty():
            print("Error: La cola está vacía")
            return None
        return self.items.pop(0)

    def front(self):
        """Retorna el primer elemento de la cola sin removerlo."""
        if self.isEmpty():
            print("Error: La cola está vacía")
            return None
        return self.items[0]
````
