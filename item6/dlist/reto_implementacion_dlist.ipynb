# 🧪 Comprobación final

````python
lista = DList()

assert lista.isEmpty()

lista.addFirst(20)
lista.addFirst(10)
lista.addLast(40)
lista.insertAt(2, 30)

assert str(lista) == "[10, 20, 30, 40]"
assert len(lista) == 4

assert lista._head.elem == 10
assert lista._tail.elem == 40
assert lista._head.prev is None
assert lista._tail.next is None

# Recorrido hacia delante.
forward = []
node = lista._head
while node:
    forward.append(node.elem)
    node = node.next

# Recorrido hacia atrás.
backward = []
node = lista._tail
while node:
    backward.append(node.elem)
    node = node.prev

assert forward == [10, 20, 30, 40]
assert backward == [40, 30, 20, 10]

assert lista.getAt(2) == 30
assert lista.index(40) == 3

assert lista.removeFirst() == 10
assert lista.removeLast() == 40
assert lista.removeAt(1) == 30

assert str(lista) == "[20]"
assert lista._head is lista._tail

assert lista.removeAt(0) == 20
assert lista.isEmpty()
assert lista._head is None
assert lista._tail is None

print("¡Todas las pruebas superadas!")
````
