# 🧱 Base code

The `DNode` class is fully implemented.

The `DList` class also provides the constructor and the basic operations for checking the size and adding elements at the beginning and end.

**Do not modify the code that is already implemented.**

````python
class DNode:
    """Node for a doubly linked list."""

    def __init__(self, elem, next=None, prev=None):
        self.elem = elem
        self.next = next
        self.prev = prev


class DList:
    """Implementation of a doubly linked list with head and tail references."""

    def __init__(self):
        # El prefijo _ indica por convención que estos atributos son internos
        # de la clase y no deberían utilizarse directamente desde fuera.
        self._head = None
        self._tail = None
        self._size = 0

    def __len__(self):
        return self._size

    def isEmpty(self):
        return len(self) == 0

    def __str__(self):
        nodeIt = self._head
        result = '['

        while nodeIt:
            if type(nodeIt.elem) == int:
                result += str(nodeIt.elem) + ", "
            else:
                result += "'" + str(nodeIt.elem) + "', "
            nodeIt = nodeIt.next

        if len(result) > 1:
            result = result[:-2]

        result += ']'
        return result

    def addFirst(self, e):
        newNode = DNode(e)

        if self.isEmpty():
            self._tail = newNode
        else:
            newNode.next = self._head
            self._head.prev = newNode

        self._head = newNode
        self._size += 1

    def addLast(self, e):
        newNode = DNode(e)

        if self.isEmpty():
            self._head = newNode
        else:
            newNode.prev = self._tail
            self._tail.next = newNode

        self._tail = newNode
        self._size += 1

    def removeFirst(self):
        result = None

        if not self.isEmpty():
            result = self._head.elem
            self._head = self._head.next

            if self._head is None:
                self._tail = None
            else:
                self._head.prev = None

            self._size -= 1

        return result

    def removeLast(self):
        result = None

        if not self.isEmpty():
            result = self._tail.elem
            self._tail = self._tail.prev

            if self._tail is None:
                self._head = None
            else:
                self._tail.next = None

            self._size -= 1

        return result

    def getAt(self, index):
        if index < 0 or index >= self._size:
            return None

        if index <= len(self) // 2:
            current = self._head
            i = 0
            while i < index:
                current = current.next
                i += 1
        else:
            current = self._tail
            i = len(self) - 1
            while i > index:
                current = current.prev
                i -= 1

        return current.elem

    def index(self, e):
        current = self._head
        idx = 0

        while current:
            if current.elem == e:
                return idx
            current = current.next
            idx += 1

        return -1

    def insertAt(self, index, e):
        if index < 0 or index > len(self):
            return

        if index == 0:
            self.addFirst(e)

        elif index == len(self):
            self.addLast(e)

        else:
            current = self._head
            for _ in range(index):
                current = current.next

            previous = current.prev
            newNode = DNode(e)

            newNode.prev = previous
            newNode.next = current

            previous.next = newNode
            current.prev = newNode

            self._size += 1

    def removeAt(self, index):
        if index < 0 or index >= len(self):
            return None

        if index == 0:
            return self.removeFirst()

        if index == len(self) - 1:
            return self.removeLast()

        current = self._head
        for _ in range(index):
            current = current.next

        previous = current.prev
        nextNode = current.next

        previous.next = nextNode
        nextNode.prev = previous

        self._size -= 1

        return current.elem
````

# 🧪 Final Check

This test combines all operations and also checks the consistency of the `next` and `prev` links.

````python
lista = DList()

assert lista.isEmpty()

lista.addFirst(20)
lista.addFirst(10)
lista.addLast(40)
lista.insertAt(2, 30)

assert str(lista) == "[10, 20, 30, 40]"
assert len(lista) == 4

assert lista._head.elem == 10
assert lista._tail.elem == 40
assert lista._head.prev is None
assert lista._tail.next is None

# Forward traversal.
forward = []
node = lista._head
while node:
    forward.append(node.elem)
    node = node.next

# Backward traversal.
backward = []
node = lista._tail
while node:
    backward.append(node.elem)
    node = node.prev

assert forward == [10, 20, 30, 40]
assert backward == [40, 30, 20, 10]

assert lista.getAt(2) == 30
assert lista.index(40) == 3

assert lista.removeFirst() == 10
assert lista.removeLast() == 40
assert lista.removeAt(1) == 30

assert str(lista) == "[20]"
assert lista._head is lista._tail

assert lista.removeAt(0) == 20
assert lista.isEmpty()
assert lista._head is None
assert lista._tail is None

print("All tests passed!")
````
