## Starting point

We will consider integers from 0 to 99, inclusive, as the possible keys.

We will assume a hash table with a fixed size of \(m = 7\) positions, indexed from 0 to 6. Although we will not implement the table, we will use a hash function to determine the index associated with each key and study possible collisions.

The following cell contains the code that implements the hash function we will use to map a key to a table index.

Remember that a collision occurs when **two different keys** have the same index (that is, the same value returned by the hash function).

````python
m = 7
KEYS = range(100)

def hash_function(key: int) -> int:
    return abs(hash(key)) % m

# Explore the hash function's output for a few keys.
for key in range(10):
    print(f"Key {key:2d} -> index {hash_function(key)}")
````

### 📝 Note: Python's `hash()` function

The `hash_function` defined above uses Python's built-in hash function internally.

Python provides the built-in function **`hash(obj)`**, which returns an integer (*hash value*) associated with an object.

For example:

```python
hash(10)      # 10
hash("hello") # An integer
```

**Important:**
- The value returned by `hash()` does not necessarily fall within the valid indices of a table.

Therefore, to obtain a valid index for a hash table of size `m`, we use the modulo operator (`%`):

```python
def hash_function(key: int) -> int:
    return abs(hash(key)) % m
```

In these challenges, we will only work with **integer keys from 0 to 99**. For these values, `hash(key)` equals the key itself, so our function is equivalent to `key % m`.
