# 📚 Before you start: how does this hash table work?

This practice uses the fundamental ideas of a hash table with **separate chaining**:

- **Table and buckets:** `table` is a list of lists. Each inner list is a bucket that can store several key-value pairs.
- **Hash function:** `_hash(key)` transforms a key into an index. The modulo operation keeps that index inside the table range.
- **Collisions:** two different keys may produce the same index.
- **Separate chaining:** colliding key-value pairs are stored together in the corresponding bucket.
- **`put(key, value)`:** inserts a new mapping or updates the value when the key already exists.
- **`get(key)`:** searches the appropriate bucket and returns the value or `None`.
- **`remove(key)`:** removes the mapping and returns its value, or `None` when the key is absent.

> **⚠️ Important**
>
> A `Map` keeps **unique keys**. Values, however, may be repeated.

# 🧱 Base code: `Map`

The following class implements a Map ADT using a fixed-capacity hash table (`7` buckets) and separate chaining.

**Run this cell before starting the challenges and do not modify it.**

````python
class Map:
    """Implementation of a Map ADT using a hash table with separate chaining."""

    def __init__(self):
        self._capacity = 7
        self.table = [[] for _ in range(self._capacity)]
        self.size = 0

    def _hash(self, key):
        """Internal hash function that computes a table position."""
        return abs(hash(key)) % self._capacity

    def put(self, key, value):
        """Adds an association or updates its value when the key already exists."""
        index = self._hash(key)
        bucket = self.table[index]

        # If the key already exists, update its value (unique keys).
        for item in bucket:
            if item[0] == key:
                item[1] = value
                return

        # If this is a new key, append [key, value].
        bucket.append([key, value])
        self.size += 1

    def get(self, key):
        """Returns the value associated with key, or None if it does not exist."""
        index = self._hash(key)
        bucket = self.table[index]

        for item in bucket:
            if item[0] == key:
                return item[1]
        return None

    def remove(self, key):
        """Removes the key-value pair associated with key."""
        index = self._hash(key)
        bucket = self.table[index]

        for i in range(len(bucket)):
            if bucket[i][0] == key:
                self.size -= 1
                return bucket.pop(i)[1]
        return None

    def __len__(self):
        return self.size

    def display(self):
        """Displays the internal structure of the table."""
        print("\n--- Map Contents ---")
        for i in range(self._capacity):
            print(f"Index {i}: {self.table[i]}")
        print(f"Total elements: {len(self)}")
        print("--------------------\n")
````
