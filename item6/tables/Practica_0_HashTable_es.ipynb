## Punto de partida

Consideraremos como posibles claves los números enteros comprendidos entre 0 y 99, ambos incluidos.

Consideraremos que la tabla hash tiene un tamaño fijo de \(m = 7\) posiciones, con índices de 0 a 6. Aunque no implementaremos la tabla, utilizaremos una función hash para determinar el índice asociado a cada clave y estudiar las posibles colisiones.

La siguiente celda contiene el código que implementa la función hash que usaremos para transformar una clave en un índice de la tabla.

Recuerda que una colisión se produce cuando **dos claves diferentes** tienen el mismo índice (es decir, el mismo valor que devuelve la función hash)

````python
m = 7
KEYS = range(100)

def hash_function(key: int) -> int:
    return abs(hash(key)) % m

# Explora el resultado de la función hash para algunas claves.
for key in range(10):
    print(f"Clave {key:2d} -> índice {hash_function(key)}")
````

### 📝 Nota: la función `hash()` de Python

La función anterior, hash_function, usa internamente la función hash de Python.

Python proporciona la función incorporada **`hash(obj)`**, que devuelve un número entero (*valor hash*) asociado a un objeto.

Por ejemplo:

```python
hash(10)      # 10
hash("hola")  # Un número entero
```

**Aspectos importantes:**
- El valor devuelto por `hash()` no está necesariamente comprendido entre los índices válidos de una tabla.

Por eso, para obtener un índice válido en una tabla hash de tamaño `m`, utilizamos el operador módulo (`%`):

```python
def hash_function(key: int) -> int:
    return abs(hash(key)) % m
```

En estos retos trabajaremos exclusivamente con **claves enteras entre 0 y 99**. Para estos, `hash(key)` coincide con la propia clave, por lo que nuestra función equivale a `key % m`.
