# 🧱 Balancing Symbols with Stacks

In this notebook, you will practise a classic application of the **Stack ADT**: checking whether a sequence of symbols is correctly balanced.

The key idea is to take advantage of the **LIFO** behaviour of a stack:

- When an opening symbol appears, push it onto the stack.
- When a closing symbol appears, it must match the most recent unmatched opening symbol.

## 📚 `Stack` class

Use this stack implementation to solve the challenges.

````python
class Stack:
    """Implementation of a LIFO stack using a native Python list."""

    def __init__(self):
        self.items = []

    def __len__(self):
        return len(self.items)

    def isEmpty(self):
        return len(self) == 0

    def push(self, e):
        self.items.append(e)

    def pop(self):
        if self.isEmpty():
            print("Error: The stack is empty")
            return None
        return self.items.pop()

    def top(self):
        if self.isEmpty():
            print("Error: The stack is empty")
            return None
        return self.items[-1]
````
