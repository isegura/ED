# 🧱 Balanceo de símbolos con pilas

En este notebook practicarás una aplicación clásica del **TAD Pila**: comprobar si una secuencia de símbolos está correctamente balanceada.

La idea fundamental es aprovechar el comportamiento **LIFO**:

- Cuando aparece un símbolo de apertura, se apila.
- Cuando aparece un símbolo de cierre, debe corresponder con el último símbolo de apertura pendiente.

## 📚 Clase `Stack`

Utiliza esta implementación de una pila para resolver los retos.

````python
class Stack:
    """Implementación de una Pila LIFO usando una lista nativa de Python."""

    def __init__(self):
        self.items = []

    def __len__(self):
        return len(self.items)

    def isEmpty(self):
        return len(self) == 0

    def push(self, e):
        self.items.append(e)

    def pop(self):
        if self.isEmpty():
            print("Error: La pila está vacía")
            return None
        return self.items.pop()

    def top(self):
        if self.isEmpty():
            print("Error: La pila está vacía")
            return None
        return self.items[-1]
````
