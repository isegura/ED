# 🧱 Código base: clase `Stack`

La siguiente clase implementa una pila LIFO utilizando internamente una lista de Python.

**Ejecuta esta celda antes de comenzar los retos y no la modifiques.**

```python
class Stack:
    """Implementación de una Pila LIFO usando una lista nativa de Python."""

    def __init__(self):
        """Inicializa una pila vacía."""
        self.items = []

    def __len__(self):
        """Retorna la cantidad de elementos en la pila."""
        return len(self.items)

    def isEmpty(self):
        """Retorna True si la pila está vacía, False en caso contrario."""
        return len(self) == 0

    def __str__(self):
        """Retorna una representación en texto de los elementos."""
        return str(self.items)

    def push(self, e):
        """Agrega el elemento e en el tope de la pila."""
        self.items.append(e)

    def pop(self):
        """Remueve y retorna el elemento en el tope de la pila."""
        if self.isEmpty():
            print("Error: La pila está vacía")
            return None
        return self.items.pop()

    def top(self):
        """Retorna el elemento del tope sin removerlo."""
        if self.isEmpty():
            print("Error: La pila está vacía")
            return None
        return self.items[-1]
```
