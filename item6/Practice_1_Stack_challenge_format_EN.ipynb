# 🧱 Base code: `Stack` class

The following class implements a LIFO stack using a Python list internally.

**Run this cell before starting the challenges and do not modify it.**

```python
class Stack:
    """Implementation of a LIFO stack using a native Python list."""

    def __init__(self):
        """Initializes an empty stack."""
        self.items = []

    def __len__(self):
        """Returns the number of elements in the stack."""
        return len(self.items)

    def isEmpty(self):
        """Returns True if the stack is empty, False otherwise."""
        return len(self) == 0

    def __str__(self):
        """Returns a string representation of the stack elements."""
        return str(self.items)

    def push(self, e):
        """Pushes element e onto the top of the stack."""
        self.items.append(e)

    def pop(self):
        """Removes and returns the element at the top of the stack."""
        if self.isEmpty():
            print("Error: The stack is empty")
            return None
        return self.items.pop()

    def top(self):
        """Returns the top element without removing it."""
        if self.isEmpty():
            print("Error: The stack is empty")
            return None
        return self.items[-1]
```
