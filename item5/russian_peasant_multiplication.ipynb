## 📜 A bit of history

**Russian peasant multiplication** is a method for multiplying two numbers using only simple operations: **halving, doubling, and addition**.

Despite its name, the underlying idea is much older. Similar multiplication methods based on **successive doubling and addition** were already used in **Ancient Egyptian mathematics**, as shown by the **Rhind Mathematical Papyrus**, written around 1650 BCE.

The method is particularly interesting from a computer science perspective: instead of performing multiplication directly, the problem is transformed through successive halvings and doublings. It is also closely related to the **binary representation of numbers**.

For example, when computing `27 × 18`, the odd values obtained while repeatedly halving 27 determine which multiples of 18 must be added.

💡 **Fun fact:** although the algorithm is very old, the ideas of **doubling, halving, and working with powers of two** are closely related to how computers represent and process numbers.
