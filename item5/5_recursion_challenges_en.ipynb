## How to Complete This Practice Session

Each challenge includes:

- 🎯 **Objective**
- 🕒 **Estimated time**
- 🟢🟡🔴 **Difficulty**
- 🧠 Design questions
- 💡 Hints

Before writing code, mentally answer:

1. What is the base case?
2. How do I reduce the problem?
3. Does each recursive call get closer to the base case?
4. What should the function return?

> - Solutions must be **recursive**.
> - You may not use `for` or `while` loops.

# ✅ Final Checklist

Always check:

- There is at least one base case.

- All recursive calls progress toward a base case.

- The recursion depth is reasonable.
