## 📜 Un poco de historia

La **multiplicación rusa**, también conocida como **multiplicación del campesino ruso** (*Russian peasant multiplication*), es un método para multiplicar dos números utilizando únicamente operaciones sencillas: **dividir entre 2, duplicar y sumar**.

Aunque su nombre hace referencia a Rusia, la idea en la que se basa es mucho más antigua. Métodos similares de multiplicación mediante **duplicaciones sucesivas y sumas** ya aparecen en las matemáticas del **Antiguo Egipto**, como muestra el **Papiro Matemático de Rhind**, escrito alrededor del 1650 a. C.

La idea resulta especialmente interesante desde el punto de vista de la informática: en lugar de realizar la multiplicación directamente, el problema se transforma mediante sucesivas divisiones y duplicaciones. Además, el método está estrechamente relacionado con la **representación binaria de los números**.

Por ejemplo, al multiplicar `27 × 18`, los valores impares que aparecen al dividir sucesivamente 27 entre 2 determinan qué múltiplos de 18 deben sumarse.

💡 **Curiosidad:** aunque el algoritmo es muy antiguo, las ideas de **duplicar, dividir entre 2 y trabajar con potencias de 2** están muy relacionadas con la forma en que los computadores representan y procesan los números.
