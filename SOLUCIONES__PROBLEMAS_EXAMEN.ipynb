# Soluciones de examen · Listas enlazadas

Implementaciones ejecutables, análisis Big O y pruebas. Se usan nodos y enlaces, **sin listas de Python para los algoritmos**. Los métodos de apoyo necesarios están implementados. Algunas ambigüedades se documentan junto a su solución.

**Convenciones:** `SList.addLast()` y `MyDList.addLast()`; en `MyDList` los métodos eliminan nodos ajustando `next`/`prev` directamente, sin `unlink()`. Cada ejercicio incluye pruebas ampliadas de contenido, tamaño y, en listas dobles, resultado y tamaño.
Las pruebas están escritas individualmente con `l = ...`, `for v in (...)`, `print(l)` y `assert`, con seis casos por problema.


## Parte I · Listas simplemente enlazadas (1–5)
Clase base para los ejercicios:

```python
class SNode:
    def __init__(self, e, next=None):
        self.elem = e
        self.next = next

class SList:
    def __init__(self):
        self._head = None
        self._size = 0

    def addLast(self, e):
        new = SNode(e)
        if self._head is None:
            self._head = new
        else:
            cur = self._head
            while cur.next is not None:
                cur = cur.next
            cur.next = new
        self._size += 1

    def __str__(self):
        cur, result = self._head, ''
        while cur is not None:
            result += (' -> ' if result else '') + str(cur.elem)
            cur = cur.next
        return result or '(vacía)'
```

### Problema 1 · `moveLastToFront(k)`

Mueve los últimos `k` nodos al principio de la lista, conservando el orden relativo. Ejemplo: `3 -> 2 -> 1 -> 7 -> 6 -> 7 -> 8`, con `k=3`, se convierte en `6 -> 7 -> 8 -> 3 -> 2 -> 1 -> 7`.

**Tareas:** implementa el método y analiza su complejidad Big O.

```python
class SList(SList):
    def moveLastToFront(self, k):
        if not 0 <= k <= self._size:
            raise ValueError('k fuera de rango')
        if k == 0 or k == self._size:
            return
        cut = self._head
        for _ in range(self._size - k - 1):
            cut = cut.next
        new_head = cut.next
        cut.next = None
        tail = new_head
        while tail.next is not None:
            tail = tail.next
        tail.next = self._head
        self._head = new_head
```

**Complejidad:** O(n) tiempo, O(1) espacio. Se localizan el corte y el último nodo mediante recorridos lineales.

```python
# PROBLEMA 1 — PRUEBAS INDIVIDUALES
# Cada prueba crea su propia lista y comprueba el resultado.

# Prueba 1: Ejemplo del examen
l = SList()
for v in (3, 2, 1, 7, 6, 7, 8):
    l.addLast(v)
result = l.moveLastToFront(3)
print(l)
assert str(l) == '6 -> 7 -> 8 -> 3 -> 2 -> 1 -> 7'
assert l._size == 7
print('✓ Prueba 1 superada')

# Prueba 2: Lista vacía
l = SList()
result = l.moveLastToFront(0)
print(l)
assert str(l) == '(vacía)'
assert l._size == 0
print('✓ Prueba 2 superada')

# Prueba 3: Lista de un elemento
l = SList()
for v in (5,):
    l.addLast(v)
result = l.moveLastToFront(0)
print(l)
assert str(l) == '5'
assert l._size == 1
print('✓ Prueba 3 superada')

# Prueba 4: Lista de un elemento
l = SList()
for v in (5,):
    l.addLast(v)
result = l.moveLastToFront(1)
print(l)
assert str(l) == '5'
assert l._size == 1
print('✓ Prueba 4 superada')

# Prueba 5: Caso adicional
l = SList()
for v in (1, 2, 3, 4):
    l.addLast(v)
result = l.moveLastToFront(2)
print(l)
assert str(l) == '3 -> 4 -> 1 -> 2'
assert l._size == 4
print('✓ Prueba 5 superada')

# Prueba 6: Caso adicional
l = SList()
for v in (1, 2, 3):
    l.addLast(v)
result = l.moveLastToFront(3)
print(l)
assert str(l) == '1 -> 2 -> 3'
assert l._size == 3
print('✓ Prueba 6 superada')

```

### Problema 2 · `deleteLast(c)`

Elimina la **última aparición** de `c`. Ejemplo: `5 -> 3 -> 2 -> 5 -> 3 -> 1`, con `c=3`, produce `5 -> 3 -> 2 -> 5 -> 1`.

**Tareas:** implementa el método y analiza su complejidad Big O.

```python
class SList(SList):
    def deleteLast(self, c):
        prev, cur = None, self._head
        last_prev, last = None, None
        while cur is not None:
            if cur.elem == c:
                last_prev, last = prev, cur
            prev, cur = cur, cur.next
        if last is None:
            return
        if last_prev is None:
            self._head = last.next
        else:
            last_prev.next = last.next
        self._size -= 1
```

**Complejidad:** O(n) tiempo, O(1) espacio: un recorrido completo.

```python
# PROBLEMA 2 — PRUEBAS INDIVIDUALES
# Cada prueba crea su propia lista y comprueba el resultado.

# Prueba 1: Ejemplo del examen
l = SList()
for v in (5, 3, 2, 5, 3, 1):
    l.addLast(v)
result = l.deleteLast(3)
print(l)
assert str(l) == '5 -> 3 -> 2 -> 5 -> 1'
assert l._size == 5
print('✓ Prueba 1 superada')

# Prueba 2: Lista vacía
l = SList()
result = l.deleteLast(3)
print(l)
assert str(l) == '(vacía)'
assert l._size == 0
print('✓ Prueba 2 superada')

# Prueba 3: Lista de un elemento
l = SList()
for v in (3,):
    l.addLast(v)
result = l.deleteLast(3)
print(l)
assert str(l) == '(vacía)'
assert l._size == 0
print('✓ Prueba 3 superada')

# Prueba 4: Lista de un elemento
l = SList()
for v in (3,):
    l.addLast(v)
result = l.deleteLast(4)
print(l)
assert str(l) == '3'
assert l._size == 1
print('✓ Prueba 4 superada')

# Prueba 5: Caso adicional
l = SList()
for v in (1, 2, 3):
    l.addLast(v)
result = l.deleteLast(1)
print(l)
assert str(l) == '2 -> 3'
assert l._size == 2
print('✓ Prueba 5 superada')

# Prueba 6: Caso adicional
l = SList()
for v in (3, 2, 3, 3):
    l.addLast(v)
result = l.deleteLast(3)
print(l)
assert str(l) == '3 -> 2 -> 3'
assert l._size == 3
print('✓ Prueba 6 superada')

```

### Problema 3 · `removeSmaller(x)`

Crea una subclase `SList2` de `SList` con un método que elimine **todos** los nodos con valores menores que `x`. No devuelve nada. **No puede llamar a otros métodos de `SList` ni utilizar listas auxiliares**.

**Tareas:** implementa el método y analiza su complejidad Big O.

```python
class SList2(SList):
    def removeSmaller(self, x):
        while self._head is not None and self._head.elem < x:
            self._head = self._head.next
            self._size -= 1
        cur = self._head
        while cur is not None and cur.next is not None:
            if cur.next.elem < x:
                cur.next = cur.next.next
                self._size -= 1
            else:
                cur = cur.next
```

**Complejidad:** O(n) tiempo y O(1) espacio.

```python
# PROBLEMA 3 — PRUEBAS INDIVIDUALES
# Cada prueba crea su propia lista y comprueba el resultado.

# Prueba 1: Ejemplo del examen
l = SList2()
for v in (2, 5, 1, 7, 3):
    l.addLast(v)
result = l.removeSmaller(4)
print(l)
assert str(l) == '5 -> 7'
assert l._size == 2
print('✓ Prueba 1 superada')

# Prueba 2: Lista vacía
l = SList2()
result = l.removeSmaller(4)
print(l)
assert str(l) == '(vacía)'
assert l._size == 0
print('✓ Prueba 2 superada')

# Prueba 3: Lista de un elemento
l = SList2()
for v in (2,):
    l.addLast(v)
result = l.removeSmaller(4)
print(l)
assert str(l) == '(vacía)'
assert l._size == 0
print('✓ Prueba 3 superada')

# Prueba 4: Lista de un elemento
l = SList2()
for v in (5,):
    l.addLast(v)
result = l.removeSmaller(4)
print(l)
assert str(l) == '5'
assert l._size == 1
print('✓ Prueba 4 superada')

# Prueba 5: Caso adicional
l = SList2()
for v in (1, 2, 3):
    l.addLast(v)
result = l.removeSmaller(4)
print(l)
assert str(l) == '(vacía)'
assert l._size == 0
print('✓ Prueba 5 superada')

# Prueba 6: Caso adicional
l = SList2()
for v in (4, 5, 6):
    l.addLast(v)
result = l.removeSmaller(4)
print(l)
assert str(l) == '4 -> 5 -> 6'
assert l._size == 3
print('✓ Prueba 6 superada')

```

### Problema 4 · `removeDoublePattern()`

Busca cadenas contiguas donde cada valor sea el doble del anterior y elimina todos los nodos de cada cadena **excepto el primero**. Devuelve el número de nodos eliminados. Ejemplo: `4, 1, 2, 4, 8, 9, 3, 2, 5, 10, 20, 2, 1, 7` → `4, 1, 9, 3, 2, 5, 2, 1, 7`; devuelve `5`. Analiza mejor y peor caso.

**Tareas:** implementa el método y analiza su complejidad Big O.

```python
class SList(SList):
    def removeDoublePattern(self):
        removed = 0
        cur = self._head
        while cur is not None:
            expected = cur.elem * 2
            nxt = cur.next
            while nxt is not None and nxt.elem == expected:
                cur.next = nxt.next
                self._size -= 1
                removed += 1
                expected = nxt.elem * 2
                nxt = cur.next
            cur = cur.next
        return removed
```

**Complejidad:** O(n) tiempo y O(1) espacio en el peor caso y, para n nodos, también O(n) en el mejor caso (se inspeccionan los nodos). Lista vacía: O(1).

```python
# PROBLEMA 4 — PRUEBAS INDIVIDUALES
# Cada prueba crea su propia lista y comprueba el resultado.

# Prueba 1: Ejemplo del examen
l = SList()
for v in (4, 1, 2, 4, 8, 9, 3, 2, 5, 10, 20, 2, 1, 7):
    l.addLast(v)
result = l.removeDoublePattern()
print(l)
assert str(l) == '4 -> 1 -> 9 -> 3 -> 2 -> 5 -> 2 -> 1 -> 7'
assert l._size == 9
print('✓ Prueba 1 superada')

# Prueba 2: Lista vacía
l = SList()
result = l.removeDoublePattern()
print(l)
assert str(l) == '(vacía)'
assert l._size == 0
assert result == 0
print('✓ Prueba 2 superada')

# Prueba 3: Lista de un elemento
l = SList()
for v in (5,):
    l.addLast(v)
result = l.removeDoublePattern()
print(l)
assert str(l) == '5'
assert l._size == 1
assert result == 0
print('✓ Prueba 3 superada')

# Prueba 4: Caso adicional
l = SList()
for v in (1, 2, 4, 8):
    l.addLast(v)
result = l.removeDoublePattern()
print(l)
assert str(l) == '1'
assert l._size == 1
assert result == 3
print('✓ Prueba 4 superada')

# Prueba 5: Caso adicional
l = SList()
for v in (1, 3, 5):
    l.addLast(v)
result = l.removeDoublePattern()
print(l)
assert str(l) == '1 -> 3 -> 5'
assert l._size == 3
assert result == 0
print('✓ Prueba 5 superada')

# Prueba 6: Caso adicional
l = SList()
for v in (0, 0, 0, 3):
    l.addLast(v)
result = l.removeDoublePattern()
print(l)
assert str(l) == '0 -> 3'
assert l._size == 2
assert result == 2
print('✓ Prueba 6 superada')

```

**Nota sobre el patrón:** se interpretan cadenas máximas consecutivas de duplicación, conservando su primer nodo.

### Problema 5 · `reverseDistinct()`

Invierte la lista y elimina todos los valores repetidos. Ejemplo: `1 -> 2 -> 3 -> 4 -> 4 -> 5 -> 6` → `6 -> 5 -> 4 -> 3 -> 2 -> 1`. Debe funcionar también con una lista vacía.

**Tareas:** implementa el método y analiza su complejidad Big O.

```python
class SList(SList):
    def reverseDistinct(self):
        prev, cur = None, self._head
        while cur is not None:
            nxt = cur.next
            cur.next = prev
            prev, cur = cur, nxt
        self._head = prev
        cur = self._head
        while cur is not None:
            runner = cur
            while runner.next is not None:
                if runner.next.elem == cur.elem:
                    runner.next = runner.next.next
                    self._size -= 1
                else:
                    runner = runner.next
            cur = cur.next
```

**Complejidad:** O(n²) tiempo en el peor caso, O(1) espacio. Se evita usar `set` o listas auxiliares.

```python
# PROBLEMA 5 — PRUEBAS INDIVIDUALES
# Cada prueba crea su propia lista y comprueba el resultado.

# Prueba 1: Ejemplo del examen
l = SList()
for v in (1, 2, 3, 4, 4, 5, 6):
    l.addLast(v)
result = l.reverseDistinct()
print(l)
assert str(l) == '6 -> 5 -> 4 -> 3 -> 2 -> 1'
assert l._size == 6
print('✓ Prueba 1 superada')

# Prueba 2: Lista vacía
l = SList()
result = l.reverseDistinct()
print(l)
assert str(l) == '(vacía)'
assert l._size == 0
print('✓ Prueba 2 superada')

# Prueba 3: Lista de un elemento
l = SList()
for v in (1,):
    l.addLast(v)
result = l.reverseDistinct()
print(l)
assert str(l) == '1'
assert l._size == 1
print('✓ Prueba 3 superada')

# Prueba 4: Caso adicional
l = SList()
for v in (1, 1, 1):
    l.addLast(v)
result = l.reverseDistinct()
print(l)
assert str(l) == '1'
assert l._size == 1
print('✓ Prueba 4 superada')

# Prueba 5: Caso adicional
l = SList()
for v in (1, 2, 3):
    l.addLast(v)
result = l.reverseDistinct()
print(l)
assert str(l) == '3 -> 2 -> 1'
assert l._size == 3
print('✓ Prueba 5 superada')

# Prueba 6: Caso adicional
l = SList()
for v in (1, 2, 1, 3, 2):
    l.addLast(v)
result = l.reverseDistinct()
print(l)
assert str(l) == '2 -> 3 -> 1'
assert l._size == 3
print('✓ Prueba 6 superada')

```

## Parte II · Listas doblemente enlazadas (6–9)
Clase base para los ejercicios:

```python
class DNode:
    def __init__(self, elem, next=None, prev=None):
        self.elem = elem
        self.next = next
        self.prev = prev

class MyDList:
    def __init__(self):
        self._head = None
        self._tail = None
        self._size = 0

    def addLast(self, e):
        node = DNode(e, prev=self._tail)
        if self._tail is None:
            self._head = node
        else:
            self._tail.next = node
        self._tail = node
        self._size += 1

    def __str__(self):
        cur, result = self._head, ''
        while cur is not None:
            result += (' <-> ' if result else '') + str(cur.elem)
            cur = cur.next
        return result or '(vacía)'
```

### Problema 6 · `removeMultiplesOf(e)`

Busca la primera aparición de `e` y elimina **después de ella** todos los valores múltiplos de `e`; no elimina la primera aparición. Si `e` no existe, muestra un mensaje. Ejemplo: `6,3,2,6,6,4,1,3,9,2` con `e=3` → `6,4,1,2`.

**Tareas:** implementa el método y analiza su complejidad Big O.

```python
class MyDList(MyDList):
    def removeMultiplesOf(self, e):
        cur = self._head
        while cur is not None and cur.elem != e:
            cur = cur.next
        if cur is None:
            print('El elemento no existe en la lista')
            return
        if e == 0:
            raise ValueError('No se puede comprobar divisibilidad por cero')
        cur = cur.next
        while cur is not None:
            nxt = cur.next
            if cur.elem % e == 0:
                anterior, siguiente = cur.prev, cur.next
                anterior.next = siguiente
                if siguiente is None:
                    self._tail = anterior
                else:
                    siguiente.prev = anterior
                self._size -= 1
            cur = nxt
```

**Complejidad:** O(n) tiempo y O(1) espacio. Para `e=0`, el enunciado no define múltiplos; aquí se señala explícitamente.

```python
# PROBLEMA 6 — PRUEBAS INDIVIDUALES
# Cada prueba crea su propia lista y comprueba el resultado.

# Prueba 1: Ejemplo del examen
l = MyDList()
for v in (6, 3, 2, 6, 6, 4, 1, 3, 9, 2):
    l.addLast(v)
result = l.removeMultiplesOf(3)
print(l)
assert str(l) == '6 <-> 3 <-> 2 <-> 4 <-> 1 <-> 2'
assert l._size == 6
print('✓ Prueba 1 superada')

# Prueba 2: Lista vacía
l = MyDList()
result = l.removeMultiplesOf(3)
print(l)
assert str(l) == '(vacía)'
assert l._size == 0
print('✓ Prueba 2 superada')

# Prueba 3: Lista de un elemento
l = MyDList()
for v in (3,):
    l.addLast(v)
result = l.removeMultiplesOf(3)
print(l)
assert str(l) == '3'
assert l._size == 1
print('✓ Prueba 3 superada')

# Prueba 4: Lista de un elemento
l = MyDList()
for v in (4,):
    l.addLast(v)
result = l.removeMultiplesOf(3)
print(l)
assert str(l) == '4'
assert l._size == 1
print('✓ Prueba 4 superada')

# Prueba 5: Caso adicional
l = MyDList()
for v in (6, 3, 2, 6, 6, 4, 1, 3, 9, 2):
    l.addLast(v)
result = l.removeMultiplesOf(1)
print(l)
assert str(l) == '6 <-> 3 <-> 2 <-> 6 <-> 6 <-> 4 <-> 1'
assert l._size == 7
print('✓ Prueba 5 superada')

# Prueba 6: Caso adicional
l = MyDList()
for v in (6, 3, 2, 6, 6, 4, 1, 3, 9, 2):
    l.addLast(v)
result = l.removeMultiplesOf(2)
print(l)
assert str(l) == '6 <-> 3 <-> 2 <-> 1 <-> 3 <-> 9'
assert l._size == 6
print('✓ Prueba 6 superada')

```

### Problema 7 · `checkSequentialEven()`

Elimina los nodos con valores impares e inserta los números pares que falten entre dos nodos consecutivos para que difieran en `2`. Ejemplo: `2,4,7,9,12,14` → `2,4,6,8,10,12,14`. Se supone que los elementos son enteros.

**Tareas:** implementa el método y analiza su complejidad Big O.

```python
class MyDList(MyDList):
    def checkSequentialEven(self):
        cur = self._head
        while cur is not None:
            nxt = cur.next
            if cur.elem % 2 != 0:
                anterior, siguiente = cur.prev, cur.next
                if anterior is None:
                    self._head = siguiente
                else:
                    anterior.next = siguiente
                if siguiente is None:
                    self._tail = anterior
                else:
                    siguiente.prev = anterior
                self._size -= 1
            cur = nxt
        cur = self._head
        while cur is not None and cur.next is not None:
            if cur.next.elem > cur.elem + 2:
                node = DNode(cur.elem + 2, next=cur.next, prev=cur)
                cur.next.prev = node
                cur.next = node
                self._size += 1
                cur = node
            else:
                cur = cur.next
```

**Complejidad:** O(n + m) tiempo, donde m es el número de nodos insertados, y O(1) espacio auxiliar (sin contar los nodos nuevos). Se presupone orden ascendente de los pares restantes, como en los ejemplos.

```python
# PROBLEMA 7 — PRUEBAS INDIVIDUALES
# Cada prueba crea su propia lista y comprueba el resultado.

# Prueba 1: Ejemplo del examen
l = MyDList()
for v in (2, 4, 7, 9, 12, 14):
    l.addLast(v)
result = l.checkSequentialEven()
print(l)
assert str(l) == '2 <-> 4 <-> 6 <-> 8 <-> 10 <-> 12 <-> 14'
assert l._size == 7
print('✓ Prueba 1 superada')

# Prueba 2: Lista vacía
l = MyDList()
result = l.checkSequentialEven()
print(l)
assert str(l) == '(vacía)'
assert l._size == 0
print('✓ Prueba 2 superada')

# Prueba 3: Lista de un elemento
l = MyDList()
for v in (3,):
    l.addLast(v)
result = l.checkSequentialEven()
print(l)
assert str(l) == '(vacía)'
assert l._size == 0
print('✓ Prueba 3 superada')

# Prueba 4: Lista de un elemento
l = MyDList()
for v in (4,):
    l.addLast(v)
result = l.checkSequentialEven()
print(l)
assert str(l) == '4'
assert l._size == 1
print('✓ Prueba 4 superada')

# Prueba 5: Caso adicional
l = MyDList()
for v in (2, 4, 6, 8):
    l.addLast(v)
result = l.checkSequentialEven()
print(l)
assert str(l) == '2 <-> 4 <-> 6 <-> 8'
assert l._size == 4
print('✓ Prueba 5 superada')

# Prueba 6: Caso adicional
l = MyDList()
for v in (2, 4, 10):
    l.addLast(v)
result = l.checkSequentialEven()
print(l)
assert str(l) == '2 <-> 4 <-> 6 <-> 8 <-> 10'
assert l._size == 5
print('✓ Prueba 6 superada')

```

### Problema 8 · `removeNeighbours(x)`

Para cada aparición de `x`, elimina sus nodos vecinos, pero **nunca** elimina un nodo cuyo valor sea `x`. Si está en un extremo, elimina solo el vecino disponible. Debe ejecutarse en O(n).

**Tareas:** implementa el método y analiza su complejidad Big O.

```python
class MyDList(MyDList):
    def removeNeighbours(self, x):
        cur = self._head
        while cur is not None:
            nxt = cur.next
            if cur.elem == x:
                if cur.prev is not None and cur.prev.elem != x:
                    doomed = cur.prev
                    anterior = doomed.prev
                    if anterior is None:
                        self._head = cur
                    else:
                        anterior.next = cur
                    cur.prev = anterior
                    self._size -= 1
                if cur.next is not None and cur.next.elem != x:
                    doomed = cur.next
                    siguiente = doomed.next
                    cur.next = siguiente
                    if siguiente is None:
                        self._tail = cur
                    else:
                        siguiente.prev = cur
                    self._size -= 1
                    nxt = siguiente
            cur = nxt
```

**Complejidad:** O(n) tiempo, O(1) espacio. Los nodos con valor x se preservan aunque sean adyacentes.

```python
# PROBLEMA 8 — PRUEBAS INDIVIDUALES
# Cada prueba crea su propia lista y comprueba el resultado.

# Prueba 1: Ejemplo del examen
l = MyDList()
for v in (9, 4, 4, 6, 3, 12, 2, 1):
    l.addLast(v)
result = l.removeNeighbours(6)
print(l)
assert str(l) == '9 <-> 4 <-> 6 <-> 12 <-> 2 <-> 1'
assert l._size == 6
print('✓ Prueba 1 superada')

# Prueba 2: Lista vacía
l = MyDList()
result = l.removeNeighbours(6)
print(l)
assert str(l) == '(vacía)'
assert l._size == 0
print('✓ Prueba 2 superada')

# Prueba 3: Lista de un elemento
l = MyDList()
for v in (6,):
    l.addLast(v)
result = l.removeNeighbours(6)
print(l)
assert str(l) == '6'
assert l._size == 1
print('✓ Prueba 3 superada')

# Prueba 4: Lista de un elemento
l = MyDList()
for v in (4,):
    l.addLast(v)
result = l.removeNeighbours(6)
print(l)
assert str(l) == '4'
assert l._size == 1
print('✓ Prueba 4 superada')

# Prueba 5: Caso adicional
l = MyDList()
for v in (6, 2, 3):
    l.addLast(v)
result = l.removeNeighbours(6)
print(l)
assert str(l) == '6 <-> 3'
assert l._size == 2
print('✓ Prueba 5 superada')

# Prueba 6: Caso adicional
l = MyDList()
for v in (2, 3, 6):
    l.addLast(v)
result = l.removeNeighbours(6)
print(l)
assert str(l) == '2 <-> 6'
assert l._size == 2
print('✓ Prueba 6 superada')

```

**Criterio para apariciones contiguas:** se conservan todos los nodos iguales a `x`, para respetar la frase «pero no el que tiene valor x».

### Problema 9 · `AddAnother(e)`

Inserta `e` inmediatamente después de su **última aparición** en la lista. Si no aparece, lo añade al final.

**Tareas:** implementa el método y analiza su complejidad Big O.

```python
class MyDList(MyDList):
    def AddAnother(self, e):
        cur = self._tail
        while cur is not None and cur.elem != e:
            cur = cur.prev
        if cur is None or cur is self._tail:
            self.addLast(e)
            return
        node = DNode(e, next=cur.next, prev=cur)
        cur.next.prev = node
        cur.next = node
        self._size += 1
```

**Complejidad:** O(n) tiempo, O(1) espacio adicional (aparte del nodo insertado).

```python
# PROBLEMA 9 — PRUEBAS INDIVIDUALES
# Cada prueba crea su propia lista y comprueba el resultado.

# Prueba 1: Ejemplo del examen
l = MyDList()
for v in (2, 3, 2, 4):
    l.addLast(v)
result = l.AddAnother(2)
print(l)
assert str(l) == '2 <-> 3 <-> 2 <-> 2 <-> 4'
assert l._size == 5
print('✓ Prueba 1 superada')

# Prueba 2: Lista vacía
l = MyDList()
result = l.AddAnother(3)
print(l)
assert str(l) == '3'
assert l._size == 1
print('✓ Prueba 2 superada')

# Prueba 3: Lista de un elemento
l = MyDList()
for v in (3,):
    l.addLast(v)
result = l.AddAnother(3)
print(l)
assert str(l) == '3 <-> 3'
assert l._size == 2
print('✓ Prueba 3 superada')

# Prueba 4: Lista de un elemento
l = MyDList()
for v in (4,):
    l.addLast(v)
result = l.AddAnother(3)
print(l)
assert str(l) == '4 <-> 3'
assert l._size == 2
print('✓ Prueba 4 superada')

# Prueba 5: Caso adicional
l = MyDList()
for v in (1, 2, 3):
    l.addLast(v)
result = l.AddAnother(3)
print(l)
assert str(l) == '1 <-> 2 <-> 3 <-> 3'
assert l._size == 4
print('✓ Prueba 5 superada')

# Prueba 6: Caso adicional
l = MyDList()
for v in (1, 2, 3):
    l.addLast(v)
result = l.AddAnother(7)
print(l)
assert str(l) == '1 <-> 2 <-> 3 <-> 7'
assert l._size == 4
print('✓ Prueba 6 superada')

```
