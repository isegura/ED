## 📚 Resumen de la implementación

La siguiente celda reúne en un único lugar la implementación básica trabajada durante la práctica: `BinaryNode`, `BinaryTree`, los cuatro recorridos (`preorder`, `inorder`, `postorder` y `levelorder`) y las funciones recursivas `size` y `height`.

Recuerda que puedes llamar a estas funciones con `tree.root` para procesar el árbol completo o con cualquier otro nodo para procesar únicamente el subárbol cuya raíz sea ese nodo.


````python
# ============================================================
# RESUMEN DEL TAD BINARY TREE
# BinaryNode, BinaryTree, traversals, size y height
# ============================================================

class BinaryNode:
    def __init__(self, e):
        self.element = e
        self.left = None
        self.right = None


class BinaryTree:
    def __init__(self):
        self.root = None


# ------------------------------------------------------------
# RECORRIDOS EN PROFUNDIDAD
# ------------------------------------------------------------

def preorder(node):
    """root - left - right"""
    if node is None:
        return []

    return (
        [node.element]
        + preorder(node.left)
        + preorder(node.right)
    )


def inorder(node):
    """left - root - right"""
    if node is None:
        return []

    return (
        inorder(node.left)
        + [node.element]
        + inorder(node.right)
    )


def postorder(node):
    """left - right - root"""
    if node is None:
        return []

    return (
        postorder(node.left)
        + postorder(node.right)
        + [node.element]
    )


# ------------------------------------------------------------
# RECORRIDO POR NIVELES
# ------------------------------------------------------------

def levelorder(node):
    """Recorrido iterativo por niveles usando una cola FIFO."""
    if node is None:
        return []

    queue = [node]
    result = []

    while queue:
        current = queue.pop(0)
        result.append(current.element)

        if current.left is not None:
            queue.append(current.left)

        if current.right is not None:
            queue.append(current.right)

    return result


# ------------------------------------------------------------
# TAMAÑO
# ------------------------------------------------------------

def size(node):
    """Devuelve el número de nodos del árbol o subárbol."""
    if node is None:
        return 0

    return 1 + size(node.left) + size(node.right)


# ------------------------------------------------------------
# ALTURA
# ------------------------------------------------------------

def height(node):
    """Devuelve la altura del árbol o subárbol.

    Convención:
        árbol vacío -> -1
        hoja        -> 0
    """
    if node is None:
        return -1

    return 1 + max(
        height(node.left),
        height(node.right)
    )
````
